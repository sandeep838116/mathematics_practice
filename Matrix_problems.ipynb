## Matrix Topic 1: Matrix Addition & Subtraction

### 1. Idea (Definition)
Matrix addition ($\mathbf{A} + \mathbf{B}$) and subtraction ($\mathbf{A} - \mathbf{B}$) take **two 2D matrices of the exact same shape** $(m, n)$ ($m$ rows and $n$ columns) and produce a **new 2D matrix of shape $(m, n)$** by adding or subtracting corresponding entries at each row $i$ and column $j$:

$$(\mathbf{A} \pm \mathbf{B})_{i,j} = A_{i,j} \pm B_{i,j} \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

> **Note:** Unlike the outer product, element-wise matrix addition and subtraction **require both matrices to have identical dimensions** ($m_A = m_B$ and $n_A = n_B$). Also, in pure Python list-of-lists form, a "jagged" input (where rows have different lengths) isn't a valid rectangular matrix—always validate shape consistency!

---

### 2. Intuition
* **Grid overlay intuition:** Think of two spreadsheets or pixel grids of the exact same width and height stacked on top of each other. Every cell $(i, j)$ combines only with the cell at the exact same position $(i, j)$—no row/column mixing happens here.
* **ML / Feature intuition:** In machine learning, element-wise matrix addition and subtraction show up in **residual / skip connections** ($\mathbf{x} + F(\mathbf{x})$ in ResNets and Transformers), **gradient updates** ($\mathbf{W} - \eta \nabla \mathbf{W}$), and **computing differences across batches or feature tables**.

---

### 3. Approach / Algorithm Behind It
1. **Validate non-empty & rectangular shape:**
   * Check that `A` and `B` are non-empty (`len(A) > 0` and `len(B) > 0`), and every row in `A` and `B` is non-empty and has a consistent row length (`len(row) == n`).
2. **Validate matching shapes:**
   * Ensure `m_A == m_B` (same number of rows) and `n_A == n_B` (same number of columns). Raise a `ValueError` if shapes differ.
3. **Allocate an $(m \times n)$ result grid:**
   * Build `res = [[0 for _ in range(n)] for _ in range(m)]` so each row is its own independent list.
4. **Nested iteration over rows and columns:**
   * Outer loop: `i` from `0` to `m - 1`.
   * Inner loop: `j` from `0` to `n - 1`, setting `res[i][j] = A[i][j] + B[i][j]` (or `-` for subtraction).
5. **Return the resulting 2D matrix:** Return `res`.

* **Time Complexity:** $O(m \cdot n)$, visiting every entry of the $m \times n$ grid once.
* **Space Complexity:** $O(m \cdot n)$ to store the resulting $m \times n$ matrix.

```python
def mat_shape_helper(mat1,mat2):
  if (not mat1) or (not mat2):
    raise ValueError("Matrix cannot be empty.")

  for i, row in enumerate(mat1):
    if not row:
      raise ValueError(f"Row {i} in matrix 1 is empty")

  for i, row in enumerate(mat2):
    if not row:
      raise ValueError(f"Row {i} in matrix 2 is empty")

  m1, n1 = len(mat1), len(mat1[0])
  m2, n2 = len(mat2), len(mat2[0])

  if any(len(row) != n1 for row in mat1):
    raise ValueError("Matrix 1 is jagged (rows have inconsistent lengths).")
  if any(len(row) != n2 for row in mat2):
    raise ValueError("Matrix 2 is jagged (rows have inconsistent lengths).")

  if m1 != m2 or n1 != n2:
    raise ValueError(
        f"Shape mismatch: ({m1}, {n1}) cannot align with ({m2}, {n2})."
    )
  return m1,n1
```

```python
def mat_add(mat1,mat2):
  m1,n1=mat_shape_helper(mat1,mat2)
  res = [[0 for _ in range(n1)] for _ in range(m1)]

  for i in range(m1):
    for j in range(n1):
      res[i][j]=mat1[i][j]+mat2[i][j]

  return res

print(mat_add([[1,2,3]],[[4,5,6]]))
```

```python
def mat_sub(mat1,mat2):
  m1,n1=mat_shape_helper(mat1,mat2)
  res = [[0 for _ in range(n1)] for _ in range(m1)]

  for i in range(m1):
    for j in range(n1):
      res[i][j]=mat1[i][j]-mat2[i][j]

  return res

print(mat_sub([[1,2,3]],[[4,5,6]]))
```

## Matrix Topic 2: Scalar Multiplication & Transpose

### 1. Idea (Definition)
This topic covers two foundational single-matrix operations on a 2D matrix $\mathbf{A}$ of shape $(m, n)$ ($m$ rows and $n$ columns):

1. **Scalar Multiplication ($\alpha \mathbf{A}$):** Multiplies every single entry of $\mathbf{A}$ by a scalar $\alpha$, producing a **new matrix of the same shape $(m, n)$**:

   $$(\alpha \mathbf{A})_{i,j} = \alpha \cdot A_{i,j} \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

2. **Matrix Transpose ($\mathbf{A}^\top$):** Flips $\mathbf{A}$ across its main diagonal by turning row $i$ into column $i$ and column $j$ into row $j$, producing a **new matrix of shape $(n, m)$**:

   $$(\mathbf{A}^\top)_{j,i} = A_{i,j} \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

> **Note:** Transpose changes shape from $(m, n)$ to $(n, m)$—a $(2, 3)$ matrix becomes $(3, 2)$. Also, always build a **new** list-of-lists rather than mutating input rows in place, especially when $m \neq n$. And just like before, validate that the matrix is non-empty with non-empty, equal-length (non-jagged) rows!

---

### 2. Intuition
* **Scalar Multiplication intuition:** Scales every entry in the grid uniformly—like turning up/down brightness on an image or scaling all weights/learning-rate updates ($\eta \nabla \mathbf{W}$) by a constant factor while keeping the layout $(m, n)$ intact.
* **Transpose intuition:** Rotates/reflects a table so rows become columns and columns become rows. In ML and linear algebra, transpose aligns shapes for multiplication ($\mathbf{X}^\top \mathbf{W}$, $\mathbf{Q}\mathbf{K}^\top$ in attention) or converts row vectors into column vectors and vice versa.

---

### 3. Approach / Algorithm Behind It

#### A. Single-Matrix Validation Helper (`_validate_matrix(mat)`)
1. Check `if not mat:` $\rightarrow$ raise `ValueError("Matrix cannot be empty.")`.
2. Check every row is non-empty (`if not row:` $\rightarrow$ raise `ValueError(...)`).
3. Set `rows = len(mat)` and `cols = len(mat[0])`; check `any(len(row) != cols for row in mat)` $\rightarrow$ raise `ValueError("Matrix is jagged.")`.
4. Return `rows, cols`.

#### B. Scalar Multiplication (`mat_scalar(mat, scalar)`)
1. Validate via `_validate_matrix(mat)` to get `rows, cols`.
2. Allocate `res = [[0 for _ in range(cols)] for _ in range(rows)]`.
3. Loop `i` in `range(rows)` and `j` in `range(cols)`: set `res[i][j] = scalar * mat[i][j]`.
4. Return `res`.

* **Time Complexity:** $O(m \cdot n)$ | **Space Complexity:** $O(m \cdot n)$

#### C. Matrix Transpose (`mat_transpose(mat)`)
1. Validate via `_validate_matrix(mat)` to get `rows, cols` (shape $(m, n)$).
2. Allocate `res = [[0 for _ in range(rows)] for _ in range(cols)]`—notice `rows` inner elements and `cols` outer rows, giving shape $(n, m)$!
3. Loop `i` in `range(rows)` and `j` in `range(cols)`: set `res[j][i] = mat[i][j]`.
4. Return `res`.

* **Time Complexity:** $O(m \cdot n)$ | **Space Complexity:** $O(n \cdot m)$ (which equals $O(m \cdot n)$ total entries)

```python
def mat_validate(mat):
  if not mat:
    raise ValueError("Matrix cannot be empty.")

  for i, row in enumerate(mat):
    if not row:
      raise ValueError(f"Row {i} in matrix is empty")


  m,n = len(mat), len(mat[0])

  if any(len(row) != n for row in mat):
    raise ValueError("Matrix is jagged (rows have inconsistent lengths).")

  return m,n
```

```python
def mat_scalar_mul(mat,num):
  m,n = mat_validate(mat)
  res =[[0 for _ in range(n)] for _ in range(m)]

  for i in range(m):
    for j in range(n):
      res[i][j] = num*mat[i][j]
  return res

print(mat_scalar_mul([[9,8,7,6,5,4,3,2,1]],8))
```

```python
def mat_transpose(mat):
  m,n = mat_validate(mat)
  res =[[0 for _ in range(m)] for _ in range(n)]

  for i in range(m):
    for j in range(n):
      res[j][i] = mat[i][j]
  return res

print(mat_transpose([[8,5,2]]))
```
