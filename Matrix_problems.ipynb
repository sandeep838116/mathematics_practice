## Matrix Topic 1: Matrix Addition & Subtraction

### 1. Idea (Definition)
Matrix addition ($\mathbf{A} + \mathbf{B}$) and subtraction ($\mathbf{A} - \mathbf{B}$) take **two 2D matrices of the exact same shape** $(m, n)$ ($m$ rows and $n$ columns) and produce a **new 2D matrix of shape $(m, n)$** by adding or subtracting corresponding entries at each row $i$ and column $j$:

$$(\mathbf{A} \pm \mathbf{B})_{i,j} = A_{i,j} \pm B_{i,j} \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

> **Note:** Unlike the outer product, element-wise matrix addition and subtraction **require both matrices to have identical dimensions** ($m_A = m_B$ and $n_A = n_B$). Also, in pure Python list-of-lists form, a "jagged" input (where rows have different lengths) isn't a valid rectangular matrix—always validate shape consistency!

---

### 2. Intuition
* **Grid overlay intuition:** Think of two spreadsheets or pixel grids of the exact same width and height stacked on top of each other. Every cell $(i, j)$ combines only with the cell at the exact same position $(i, j)$—no row/column mixing happens here.
* **ML / Feature intuition:** In machine learning, element-wise matrix addition and subtraction show up in **residual / skip connections** ($\mathbf{x} + F(\mathbf{x})$ in ResNets and Transformers), **gradient updates** ($\mathbf{W} - \eta \nabla \mathbf{W}$), and **computing differences across batches or feature tables**.

---

### 3. Approach / Algorithm Behind It
1. **Validate non-empty & rectangular shape:**
   * Check that `A` and `B` are non-empty (`len(A) > 0` and `len(B) > 0`), and every row in `A` and `B` is non-empty and has a consistent row length (`len(row) == n`).
2. **Validate matching shapes:**
   * Ensure `m_A == m_B` (same number of rows) and `n_A == n_B` (same number of columns). Raise a `ValueError` if shapes differ.
3. **Allocate an $(m \times n)$ result grid:**
   * Build `res = [[0 for _ in range(n)] for _ in range(m)]` so each row is its own independent list.
4. **Nested iteration over rows and columns:**
   * Outer loop: `i` from `0` to `m - 1`.
   * Inner loop: `j` from `0` to `n - 1`, setting `res[i][j] = A[i][j] + B[i][j]` (or `-` for subtraction).
5. **Return the resulting 2D matrix:** Return `res`.

* **Time Complexity:** $O(m \cdot n)$, visiting every entry of the $m \times n$ grid once.
* **Space Complexity:** $O(m \cdot n)$ to store the resulting $m \times n$ matrix.

```python
def mat_shape_helper(mat1,mat2):
  if (not mat1) or (not mat2):
    raise ValueError("Matrix cannot be empty.")

  for i, row in enumerate(mat1):
    if not row:
      raise ValueError(f"Row {i} in matrix 1 is empty")

  for i, row in enumerate(mat2):
    if not row:
      raise ValueError(f"Row {i} in matrix 2 is empty")

  m1, n1 = len(mat1), len(mat1[0])
  m2, n2 = len(mat2), len(mat2[0])

  if any(len(row) != n1 for row in mat1):
    raise ValueError("Matrix 1 is jagged (rows have inconsistent lengths).")
  if any(len(row) != n2 for row in mat2):
    raise ValueError("Matrix 2 is jagged (rows have inconsistent lengths).")

  if m1 != m2 or n1 != n2:
    raise ValueError(
        f"Shape mismatch: ({m1}, {n1}) cannot align with ({m2}, {n2})."
    )
  return m1,n1
```

```python
def mat_add(mat1,mat2):
  m1,n1=mat_shape_helper(mat1,mat2)
  res = [[0 for _ in range(n1)] for _ in range(m1)]

  for i in range(m1):
    for j in range(n1):
      res[i][j]=mat1[i][j]+mat2[i][j]

  return res

print(mat_add([[1,2,3]],[[4,5,6]]))
```

```python
def mat_sub(mat1,mat2):
  m1,n1=mat_shape_helper(mat1,mat2)
  res = [[0 for _ in range(n1)] for _ in range(m1)]

  for i in range(m1):
    for j in range(n1):
      res[i][j]=mat1[i][j]-mat2[i][j]

  return res

print(mat_sub([[1,2,3]],[[4,5,6]]))
```

## Matrix Topic 2: Scalar Multiplication & Transpose

### 1. Idea (Definition)
This topic covers two foundational single-matrix operations on a 2D matrix $\mathbf{A}$ of shape $(m, n)$ ($m$ rows and $n$ columns):

1. **Scalar Multiplication ($\alpha \mathbf{A}$):** Multiplies every single entry of $\mathbf{A}$ by a scalar $\alpha$, producing a **new matrix of the same shape $(m, n)$**:

   $$(\alpha \mathbf{A})_{i,j} = \alpha \cdot A_{i,j} \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

2. **Matrix Transpose ($\mathbf{A}^\top$):** Flips $\mathbf{A}$ across its main diagonal by turning row $i$ into column $i$ and column $j$ into row $j$, producing a **new matrix of shape $(n, m)$**:

   $$(\mathbf{A}^\top)_{j,i} = A_{i,j} \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

> **Note:** Transpose changes shape from $(m, n)$ to $(n, m)$—a $(2, 3)$ matrix becomes $(3, 2)$. Also, always build a **new** list-of-lists rather than mutating input rows in place, especially when $m \neq n$. And just like before, validate that the matrix is non-empty with non-empty, equal-length (non-jagged) rows!

---

### 2. Intuition
* **Scalar Multiplication intuition:** Scales every entry in the grid uniformly—like turning up/down brightness on an image or scaling all weights/learning-rate updates ($\eta \nabla \mathbf{W}$) by a constant factor while keeping the layout $(m, n)$ intact.
* **Transpose intuition:** Rotates/reflects a table so rows become columns and columns become rows. In ML and linear algebra, transpose aligns shapes for multiplication ($\mathbf{X}^\top \mathbf{W}$, $\mathbf{Q}\mathbf{K}^\top$ in attention) or converts row vectors into column vectors and vice versa.

---

### 3. Approach / Algorithm Behind It

#### A. Single-Matrix Validation Helper (`_validate_matrix(mat)`)
1. Check `if not mat:` $\rightarrow$ raise `ValueError("Matrix cannot be empty.")`.
2. Check every row is non-empty (`if not row:` $\rightarrow$ raise `ValueError(...)`).
3. Set `rows = len(mat)` and `cols = len(mat[0])`; check `any(len(row) != cols for row in mat)` $\rightarrow$ raise `ValueError("Matrix is jagged.")`.
4. Return `rows, cols`.

#### B. Scalar Multiplication (`mat_scalar(mat, scalar)`)
1. Validate via `_validate_matrix(mat)` to get `rows, cols`.
2. Allocate `res = [[0 for _ in range(cols)] for _ in range(rows)]`.
3. Loop `i` in `range(rows)` and `j` in `range(cols)`: set `res[i][j] = scalar * mat[i][j]`.
4. Return `res`.

* **Time Complexity:** $O(m \cdot n)$ | **Space Complexity:** $O(m \cdot n)$

#### C. Matrix Transpose (`mat_transpose(mat)`)
1. Validate via `_validate_matrix(mat)` to get `rows, cols` (shape $(m, n)$).
2. Allocate `res = [[0 for _ in range(rows)] for _ in range(cols)]`—notice `rows` inner elements and `cols` outer rows, giving shape $(n, m)$!
3. Loop `i` in `range(rows)` and `j` in `range(cols)`: set `res[j][i] = mat[i][j]`.
4. Return `res`.

* **Time Complexity:** $O(m \cdot n)$ | **Space Complexity:** $O(n \cdot m)$ (which equals $O(m \cdot n)$ total entries)

```python
def mat_validate(mat):
  if not mat:
    raise ValueError("Matrix cannot be empty.")

  for i, row in enumerate(mat):
    if not row:
      raise ValueError(f"Row {i} in matrix is empty")


  m,n = len(mat), len(mat[0])

  if any(len(row) != n for row in mat):
    raise ValueError("Matrix is jagged (rows have inconsistent lengths).")

  return m,n
```

```python
def mat_scalar_mul(mat,num):
  m,n = mat_validate(mat)
  res =[[0 for _ in range(n)] for _ in range(m)]

  for i in range(m):
    for j in range(n):
      res[i][j] = num*mat[i][j]
  return res

print(mat_scalar_mul([[9,8,7,6,5,4,3,2,1]],8))
```

```python
def mat_transpose(mat):
  m,n = mat_validate(mat)
  res =[[0 for _ in range(m)] for _ in range(n)]

  for i in range(m):
    for j in range(n):
      res[j][i] = mat[i][j]
  return res

print(mat_transpose([[8,5,2]]))
```

## Matrix Topic 3: Matrix-Vector Multiplication ($\mathbf{A}\mathbf{x}$)

### 1. Idea (Definition)
Matrix-vector multiplication takes a 2D matrix $\mathbf{A}$ of shape $(m, n)$ ($m$ rows and $n$ columns) and a 1D vector $\mathbf{x}$ of length $n$, and produces a **1D output vector $\mathbf{y}$ of length $m$**.

Each output component $y_i$ (for row index $i \in \{0, 1, \dots, m-1\}$) is the **dot product** of the $i$-th row of $\mathbf{A}$ with the vector $\mathbf{x}$:

$$y_i = (\mathbf{A}\mathbf{x})_i = \sum_{j=0}^{n-1} A_{i,j}\, x_j = A_{i,0}x_0 + A_{i,1}x_1 + \dots + A_{i,n-1}x_{n-1}$$

> **Note on Shape Compatibility:** The number of **columns** in $\mathbf{A}$ ($n$) must equal the **length** of vector $\mathbf{x}$ ($n$):
> $$(m, \mathbf{n}) \times (\mathbf{n},) \longrightarrow (m,)$$
> If `n != len(x)` or `x` is empty, raise a `ValueError`. Also notice the output is a **1D list of length $m$** (not an $m \times 1$ nested list).

---

### 2. Intuition
* **Linear combination of columns / bank of dot products:**
  * **Row-dot-product view:** Every row of $\mathbf{A}$ acts like a weight vector/filter, and $y_i$ measures how strongly row $i$ aligns with $\mathbf{x}$.
  * **Column linear-combination view:** $\mathbf{A}\mathbf{x} = x_0\,\mathbf{a}_{:,0} + x_1\,\mathbf{a}_{:,1} + \dots + x_{n-1}\,\mathbf{a}_{:,n-1}$—the entries of $\mathbf{x}$ act as weights scaling each column of $\mathbf{A}$ and summing them up.
* **ML / Feature intuition:** A single dense/linear layer forward pass (before adding bias) is literally $\mathbf{z} = \mathbf{W}\mathbf{x}$ (or $\mathbf{x}^\top\mathbf{W}$), mapping an $n$-dimensional input feature vector into an $m$-dimensional output representation.

---

### 3. Approach / Algorithm Behind It
1. **Validate matrix $\mathbf{A}$:** Call `m, n = mat_validate(mat)` so $\mathbf{A}$ is guaranteed non-empty, has non-empty rows, and has uniform column width `n`.
2. **Validate vector $\mathbf{x}$:**
   * Check `if not vec:` $\rightarrow$ raise `ValueError("Vector cannot be empty.")`.
   * Check `if len(vec) != n:` $\rightarrow$ raise `ValueError(f"Dimension mismatch: matrix has {n} columns, but vector has length {len(vec)}.")`.
3. **Allocate 1D result vector of length `m`:** Initialize `res = [0 for _ in range(m)]`.
4. **Compute dot product for each row:**
   * Outer loop: `i` from `0` to `m - 1`.
   * Initialize `row_sum = 0`.
   * Inner loop: `j` from `0` to `n - 1`, accumulating `row_sum += mat[i][j] * vec[j]`.
   * Store `res[i] = row_sum`.
5. **Return `res`:** Return the 1D list of length `m`.

* **Time Complexity:** $O(m \cdot n)$, performing $m \cdot n$ multiplications and additions across all rows and columns.
* **Space Complexity:** $O(m)$ for the output vector of length $m$ (plus $O(1)$ extra working variables).

```python
def mat_vec_mul(mat,vec):
  m,n = mat_validate(mat)
  if not vec:
    raise ValueError("Vector cannot be empty.")
  if len(vec)!=n:
    raise ValueError(f"Dimension mismatched: matrix has {n} columns, but vector has length {len(vec)}.")

  res = [0 for _ in range(m)]

  for i in range(m):
    row_sum =0
    for j in range(n):
      row_sum += (mat[i][j]*vec[j])
    res[i] = row_sum

  return res

print(mat_vec_mul([[1,2,3,4,5]],[1,4,7,2,5]))
```

## Matrix Topic 4: Matrix-Matrix Multiplication ($\mathbf{A}\mathbf{B}$)

### 1. Idea (Definition)
Matrix-matrix multiplication takes a 2D matrix $\mathbf{A}$ of shape $(m, k)$ ($m$ rows and $k$ columns) and a 2D matrix $\mathbf{B}$ of shape $(k, n)$ ($k$ rows and $n$ columns), and produces a **2D output matrix $\mathbf{C} = \mathbf{A}\mathbf{B}$ of shape $(m, n)$**.

Each entry $C_{i,j}$ (for row $i \in \{0, \dots, m-1\}$ and column $j \in \{0, \dots, n-1\}$) is the **dot product of the $i$-th row of $\mathbf{A}$ with the $j$-th column of $\mathbf{B}$**:

$$C_{i,j} = (\mathbf{A}\mathbf{B})_{i,j} = \sum_{p=0}^{k-1} A_{i,p}\, B_{p,j} = A_{i,0}B_{0,j} + A_{i,1}B_{1,j} + \dots + A_{i,k-1}B_{k-1,j}$$

> **Note on Shape Compatibility:** The inner dimensions must match—the number of **columns in $\mathbf{A}$** ($k$) must equal the number of **rows in $\mathbf{B}$** ($k$):
> $$(m, \mathbf{k}) \times (\mathbf{k}, n) \longrightarrow (m, n)$$
> If `n_A != m_B`, raise a `ValueError`. Also remember: matrix multiplication is **not commutative** in general ($\mathbf{A}\mathbf{B} \neq \mathbf{B}\mathbf{A}$), even when both products are valid!

---

### 2. Intuition
* **Grid of dot products:** Think of $\mathbf{C}$ as an $(m \times n)$ scorecard where cell $(i, j)$ answers: *"What is the dot product between row $i$ of $\mathbf{A}$ and column $j$ of $\mathbf{B}$?"*
* **Batching matrix-vector products:** Multiplying $\mathbf{A}$ $(m, k)$ by $\mathbf{B}$ $(k, n)$ is equivalent to multiplying $\mathbf{A}$ by each column of $\mathbf{B}$ one by one and placing those resulting column vectors side by side as the columns of $\mathbf{C}$.
* **ML / Feature intuition:** A mini-batch linear layer $\mathbf{Y} = \mathbf{X}\mathbf{W}$ (where $\mathbf{X}$ has shape `(batch_size, in_features)` and $\mathbf{W}$ has shape `(in_features, out_features)`) or scaled dot-product attention scores $\mathbf{Q}\mathbf{K}^\top$ is built directly on this operation.

---

### 3. Approach / Algorithm Behind It
1. **Validate both matrices:**
   * Call `m1, n1 = mat_validate(mat1)` and `m2, n2 = mat_validate(mat2)`.
   * Here, `m1` is rows of `A`, `n1` (which equals `k`) is columns of `A`, `m2` is rows of `B`, and `n2` is columns of `B`.
2. **Validate inner dimension alignment:**
   * Check `if n1 != m2:` $\rightarrow$ raise `ValueError(f"Dimension mismatch: mat1 has {n1} columns, but mat2 has {m2} rows.")`.
3. **Allocate an $(m_1 \times n_2)$ result matrix:**
   * Initialize `res = [[0 for _ in range(n2)] for _ in range(m1)]`.
4. **Triple-nested loop (`i`, `j`, `p`):**
   * Outer loop: `i` from `0` to `m1 - 1` (row of `mat1`).
   * Middle loop: `j` from `0` to `n2 - 1` (column of `mat2`).
   * Inner loop: `p` from `0` to `n1 - 1` (shared inner dimension `k`), accumulating:
     `dot_sum += mat1[i][p] * mat2[p][j]`
   * Assign `res[i][j] = dot_sum`.
5. **Return `res`:** Return the $(m_1 \times n_2)$ list of lists.

* **Time Complexity:** $O(m_1 \cdot n_1 \cdot n_2)$, performing $m_1 \cdot n_1 \cdot n_2$ scalar multiplications and additions.
* **Space Complexity:** $O(m_1 \cdot n_2)$ for the output matrix (plus $O(1)$ extra working variables).

```python
def mat_mat_mul(mat1,mat2):
  m1,n1 = mat_validate(mat1)
  m2,n2 = mat_validate(mat2)

  if n1 != m2 :
    raise ValueError(f"Dimension mismatch: mat1 has {n1} columns, but mat2 has {m2} rows.")

  res = [[0 for _ in range(n2)] for _ in range(m1)]

  for i in range(m1):
    for j in range(n2):
      dot_sum =0
      for p in range(n1):
        dot_sum +=(mat1[i][p]*mat2[p][j])
      res[i][j]=dot_sum
  return res

print(mat_mat_mul([[1,2,3]],[[1],[2],[3]]))
```

## Matrix Topic 5: Matrix Trace & Frobenius Norm

### 1. Idea (Definition)
This topic introduces two classic scalar summaries of a 2D matrix $\mathbf{A}$:

1. **Matrix Trace ($\operatorname{tr}(\mathbf{A})$):** Defined for a **square matrix** $\mathbf{A}$ of shape $(n, n)$, the trace is the **sum of the entries along the main diagonal** (where row index equals column index):

   $$\operatorname{tr}(\mathbf{A}) = \sum_{i=0}^{n-1} A_{i,i} = A_{0,0} + A_{1,1} + \dots + A_{n-1,n-1}$$

2. **Frobenius Norm ($\|\mathbf{A}\|_F$):** Defined for **any 2D matrix** $\mathbf{A}$ of shape $(m, n)$, the Frobenius norm is the **square root of the sum of squares of all its entries** (just like stretching all $m \cdot n$ entries into a single 1D vector and taking its $L_2$ norm):

   $$\|\mathbf{A}\|_F = \sqrt{\sum_{i=0}^{m-1} \sum_{j=0}^{n-1} A_{i,j}^2}$$

> **Note:** Trace requires a **square** matrix ($m == n$); if a matrix isn't square (e.g., shape `(2, 3)`), raise a `ValueError`. By contrast, the Frobenius norm works on **any valid rectangular** $(m, n)$ matrix—including non-square shapes.

---

### 2. Intuition
* **Trace intuition:** Summarizes the main diagonal of a square matrix in a single number. In linear algebra and ML, $\operatorname{tr}(\mathbf{A})$ equals the **sum of the eigenvalues** of $\mathbf{A}$, and shows up in **covariance matrix analysis** (total variance across features) and **matrix calculus** identities.
* **Frobenius norm intuition:** Measures the overall "size" or "energy" of a whole 2D table/matrix—equivalent to flattening the $(m \times n)$ grid into a vector of length $m \cdot n$ and computing its Euclidean ($L_2$) length. In ML, $\|\mathbf{W}\|_F^2$ appears in **weight decay / $L_2$ regularization**, and $\|\mathbf{A} - \hat{\mathbf{A}}\|_F$ measures **matrix reconstruction error**.

---

### 3. Approach / Algorithm Behind It

#### A. Matrix Trace (`mat_trace(mat)`)
1. Validate via `m, n = mat_validate(mat)`.
2. Check `if m != n:` $\rightarrow$ raise `ValueError(f"Trace requires a square matrix, got shape ({m}, {n}).")`.
3. Initialize `trace_sum = 0`.
4. Loop `i` from `0` to `n - 1`: accumulate `trace_sum += mat[i][i]`.
5. Return `trace_sum`.

* **Time Complexity:** $O(n)$ (only visits the $n$ diagonal elements of an $n \times n$ matrix).
* **Space Complexity:** $O(1)$ extra space.

#### B. Frobenius Norm (`mat_frobenius_norm(mat)`)
1. Validate via `m, n = mat_validate(mat)`.
2. Initialize `sq_sum = 0`.
3. Outer loop `i` from `0` to `m - 1`, inner loop `j` from `0` to `n - 1`: accumulate `sq_sum += mat[i][j] ** 2` (or `mat[i][j] * mat[i][j]`).
4. Return `math.sqrt(sq_sum)`.

* **Time Complexity:** $O(m \cdot n)$ (visits every entry in the $m \times n$ matrix).
* **Space Complexity:** $O(1)$ extra space.

```python
def mat_trace(mat):
  m,n = mat_validate(mat)
  if m!=n:
    raise ValueError(f"Trace requires a square matrix, got shape ({m},{n}).")

  trace =0
  for i in range(n):
    trace += mat[i][i]
  return trace

print(mat_trace([[1,20],[4,5]]))
```

```python
import math
def mat_frobenius_norm(mat):
  m,n = mat_validate(mat)

  frob_sum =0
  for i in range(m):
    for j in range(n):
      frob_sum += (mat[i][j]**2)

  return math.sqrt(frob_sum)

print(mat_frobenius_norm([[7,5,3,2,6,9,4,1]]))
```

## Matrix Topic 6: Special Matrices (Identity Matrix & Diagonal Extraction)

### 1. Idea (Definition)
This topic covers two classic structural operations on 2D matrices:

1. **Identity Matrix ($\mathbf{I}_n$):** A **square matrix** of shape $(n, n)$ with **`1` along the main diagonal** ($i = j$) and **`0` everywhere else** ($i \neq j$):

   $$(\mathbf{I}_n)_{i,j} = \begin{cases} 1 & \text{if } i = j \\ 0 & \text{if } i \neq j \end{cases}$$

   For example, $\mathbf{I}_3$ is:
   $$\mathbf{I}_3 = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{bmatrix}$$

2. **Diagonal Extraction (`mat_diag`):** Given a 2D matrix $\mathbf{A}$ of shape $(m, n)$, extracts the **main diagonal** as a **1D list** of length $\min(m, n)$:

   $$d_k = A_{k,k} \quad \text{for } 0 \le k < \min(m, n)$$

> **Note:** An identity matrix is only defined for a positive integer size $n \ge 1$ (`if n <= 0:` raise `ValueError`). Diagonal extraction works on **both square and non-square** validated matrices—for example, a `(2, 3)` or `(3, 2)` matrix has a main diagonal of length `2` (`min(m, n)`).

---

### 2. Intuition
* **Identity matrix intuition:** The multiplicative identity for matrices—just like `1` in ordinary multiplication, $\mathbf{A}\mathbf{I}_n = \mathbf{A}$ and $\mathbf{I}_m\mathbf{A} = \mathbf{A}$ (whenever shapes align). In linear algebra and ML, $\mathbf{I}$ is the baseline for **matrix inverses** ($\mathbf{A}^{-1}\mathbf{A} = \mathbf{I}$), **orthogonal matrices** ($\mathbf{Q}^\top\mathbf{Q} = \mathbf{I}$), and **ridge regularization / Tikhonov terms** ($\lambda\mathbf{I}$).
* **Diagonal extraction intuition:** Pulls out the "self-alignment" or "same-index" entries of a matrix. Notice how `sum(mat_diag(A))` for a square matrix gives the exact same value as `mat_trace(A)`!

---

### 3. Approach / Algorithm Behind It

#### A. Identity Matrix (`mat_identity(n)`)
1. Validate size: check `if not isinstance(n, int) or isinstance(n, bool) or n <= 0:` $\rightarrow$ raise `ValueError("Size n must be a positive integer.")`.
2. Allocate an $(n \times n)$ matrix filled with `0`:
   `res = [[0 for _ in range(n)] for _ in range(n)]`
3. Set main diagonal entries to `1`:
   Loop `i` from `0` to `n - 1`: set `res[i][i] = 1`.
4. Return `res`.

* **Time Complexity:** $O(n^2)$ to construct the $n \times n$ grid.
* **Space Complexity:** $O(n^2)$ for the output matrix.

#### B. Diagonal Extraction (`mat_diag(mat)`)
1. Validate via `m, n = mat_validate(mat)`.
2. Compute `diag_len = min(m, n)`.
3. Extract `res = [mat[i][i] for i in range(diag_len)]` (or via a `for i in range(diag_len):` loop).
4. Return `res`.

* **Time Complexity:** $O(\min(m, n))$.
* **Space Complexity:** $O(\min(m, n))$ for the 1D list of diagonal entries.

```python
def mat_identity(n):
  if not isinstance(n, int) or isinstance(n, bool) or n <= 0:
    raise ValueError(f"Size {n} must be a positive number.")

  res =[[0 for _ in range(n)] for _ in range(n)]

  for i in range(n):
    res[i][i]=1
  return res

print(mat_identity(5))
```

```python
def mat_diag(mat):
  m,n = mat_validate(mat)
  diag_len = min(m,n)

  res =[mat[i][i] for i in range(diag_len)]
  return res

print(mat_diag([[7,8,9],[8,5,2]]))
```

## Matrix Topic 7: Hadamard Product & Outer Product ($\mathbf{u}\mathbf{v}^\top$)

### 1. Idea (Definition)
This topic introduces two element-wise / outer-structure matrix operations widely used in linear algebra and machine learning:

1. **Hadamard Product ($\mathbf{A} \odot \mathbf{B}$):** For two matrices $\mathbf{A}$ and $\mathbf{B}$ of the **exact same shape** $(m, n)$, the Hadamard (element-wise) product produces a **new matrix of shape $(m, n)$** by multiplying corresponding entries:

   $$(\mathbf{A} \odot \mathbf{B})_{i,j} = A_{i,j} \cdot B_{i,j} \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

2. **Outer Product ($\mathbf{u} \otimes \mathbf{v}$ or $\mathbf{u}\mathbf{v}^\top$):** Given two non-empty 1D vectors $\mathbf{u}$ of length $m$ and $\mathbf{v}$ of length $n$, the outer product produces a **2D matrix $\mathbf{M}$ of shape $(m, n)$** where entry $(i, j)$ is the product of $u_i$ and $v_j$:

   $$M_{i,j} = u_i \cdot v_j \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

> **Note:** Don't confuse **Hadamard product** ($\mathbf{A} \odot \mathbf{B}$) with **matrix multiplication** ($\mathbf{A}\mathbf{B}$): Hadamard requires **identical shapes** $(m, n)$ and multiplies entry-by-entry, whereas standard matrix multiplication requires $(m, k) \times (k, n) \to (m, n)$. And unlike the inner (dot) product of two vectors—which produces a **scalar**—the **outer product** of a length-$m$ vector and a length-$n$ vector produces an **$(m \times n)$ matrix** (with rank at most 1).

---

### 2. Intuition
* **Hadamard product intuition:** Acts like a mask, gate, or element-wise scaling filter on a 2D grid. In neural networks, gating mechanisms (such as LSTM gates or SwiGLU / gated activations) rely directly on element-wise multiplication ($\odot$).
* **Outer product intuition:** Expands two 1D vectors into a full 2D interaction grid—every element of $\mathbf{u}$ pairs with every element of $\mathbf{v}$. In ML and linear algebra, outer products build **rank-1 matrices**, **covariance/Gram terms** ($\mathbf{x}\mathbf{x}^\top$), and **weight updates** ($\delta \mathbf{W} = \eta\, \boldsymbol{\delta}\mathbf{x}^\top$ in backpropagation).

---

### 3. Approach / Algorithm Behind It

#### A. Hadamard Product (`mat_hadamard(mat1, mat2)`)
1. Validate both matrices and check equal shapes using your existing `m1, n1 = mat_shape_helper(mat1, mat2)`.
2. Allocate `res = [[0 for _ in range(n1)] for _ in range(m1)]`.
3. Loop `i` from `0` to `m1 - 1` and `j` from `0` to `n1 - 1`:
   set `res[i][j] = mat1[i][j] * mat2[i][j]`.
4. Return `res`.

* **Time Complexity:** $O(m \cdot n)$ | **Space Complexity:** $O(m \cdot n)$

#### B. Outer Product (`vec_outer(vec1, vec2)`)
1. Validate both 1D vectors:
   * Check `if not vec1 or not vec2:` $\rightarrow$ raise `ValueError("Vectors cannot be empty.")`.
2. Let `m = len(vec1)` and `n = len(vec2)`.
3. Allocate `res = [[0 for _ in range(n)] for _ in range(m)]` (note: `m` rows, `n` columns).
4. Loop `i` from `0` to `m - 1` and `j` from `0` to `n - 1`:
   set `res[i][j] = vec1[i] * vec2[j]`.
5. Return `res`.

* **Time Complexity:** $O(m \cdot n)$ | **Space Complexity:** $O(m \cdot n)$

```python
def mat_hadamard(mat1,mat2):
  m,n = mat_shape_helper(mat1,mat2)
  res =[[0 for _ in range(n)] for _ in range(m)]
  for i in range(m):
    for j in range(n):
      res[i][j] = mat1[i][j] * mat2[i][j]

  return res

print(mat_hadamard([[1,1,2,3,4,5]],[[7,4,1,8,5,2]]))
```

```python
def vec_outer(vec1,vec2):
  if not vec1 or not vec2:
    raise ValueError("Vectors cannot be empty.")
  m,n = len(vec1),len(vec2)
  res = [[0 for _ in range(n)]for _ in range(m)]
  for i in range(m):
    for j in range(n):
      res[i][j]= vec1[i]*vec2[j]

  return res

print(vec_outer([1,2,3,4,5,6],[4,5,6,7,8,9]))
```

## Matrix Topic 8: Symmetry Checks & Row / Column Reductions

### 1. Idea (Definition)
This topic introduces structural symmetry checks on square matrices alongside row-wise and column-wise reductions on 2D matrices:

1. **Symmetric & Skew-Symmetric Checks:** For a **square matrix** $\mathbf{A}$ of shape $(n, n)$:
   * **Symmetric ($\mathbf{A}^\top = \mathbf{A}$):** Every off-diagonal entry mirrors its transpose partner:
     $$A_{i,j} = A_{j,i} \quad \text{for all } 0 \le i, j < n$$
   * **Skew-Symmetric / Antisymmetric ($\mathbf{A}^\top = -\mathbf{A}$):** Every entry equals the negative of its transpose partner (which also forces main-diagonal entries to be $0$):
     $$A_{i,j} = -A_{j,i} \quad \text{for all } 0 \le i, j < n$$

2. **Row Sums & Column Sums:** Given a validated 2D matrix $\mathbf{A}$ of shape $(m, n)$:
   * **Row sums:** Produces a **1D list of length $m$** containing the sum of entries in each row $i$.
   * **Column sums:** Produces a **1D list of length $n$** containing the sum of entries in each column $j$.

> **Note:** Symmetry and skew-symmetry are defined on **square** matrices ($m == n$); non-square matrices should raise a `ValueError` rather than silently returning `False`. By contrast, row and column sums work on **any valid rectangular** $(m, n)$ matrix.

---

### 2. Intuition
* **Symmetry & skew-symmetry intuition:** Symmetry means flipping across the main diagonal changes nothing—like covariance/Gram matrices or adjacency matrices of undirected graphs. Skew-symmetry flips signs across the diagonal (and zeros out the diagonal itself), showing up in rotational/infinitesimal generators and matrix decompositions ($\mathbf{A} = \frac{1}{2}(\mathbf{A} + \mathbf{A}^\top) + \frac{1}{2}(\mathbf{A} - \mathbf{A}^\top)$).
* **Row & column sum intuition:** Collapses a 2D table along one axis at a time—such as summing across features for each sample vs. summing across samples for each feature, or computing total row/column weights in probability/transition tables.

---

### 3. Approach / Algorithm Behind It

* **Symmetry & skew-symmetry checks:** Start by validating the matrix and ensuring it is square. Think about which entries actually need comparison (avoiding redundant work) and whether floating-point tolerances or exact equality make sense for your inputs. Both run in $O(n^2)$ time and $O(1)$ extra space.
* **Row & column sums:** Validate the matrix first so dimensions are trustworthy. Decide which output axis you are building (length $m$ vs. length $n$) and how you want to accumulate across the opposite dimension. Both run in $O(m \cdot n)$ time with $O(m)$ or $O(n)$ output space.

```python
def mat_sym(mat):
  m,n = mat_validate(mat)
  if m!=n:
    raise ValueError("Matrix should be square.")

  for i in range(m):
    for j in range(i+1,n):
      if mat[j][i] != mat[i][j]:
        return False
  return True
print(mat_sym([[1,20],[20,1]]))
```

```python
def mat_skew_sym(mat):
  m,n = mat_validate(mat)
  if m!=n:
    raise ValueError("Matrix should be square.")

  for i in range(m):
    if mat[i][i] != 0:
      return False
    for j in range(i+1,n):
      if mat[j][i] != -mat[i][j]:
        return False
  return True
print(mat_skew_sym([[1,20],[20,1]]))
```

```python
def mat_col_sum(mat):
  m,n = mat_validate(mat)
  res =[0 for _ in range(n)]
  for i in range(n):
    row_sum =0
    for j in range(m):
      row_sum+=mat[j][i]
    res[i]=row_sum
  return res

print(mat_col_sum([[1,2,8,4,5],[6,8,7,6,5]]))
```

```python
def mat_row_sum(mat):
  m,n = mat_validate(mat)
  res =[0 for _ in range(m)]
  for i in range(m):
    row_sum =0
    for j in range(n):
      row_sum+=mat[i][j]
    res[i]=row_sum
  return res

print(mat_row_sum([[1,2,8,4,5],[6,8,7,6,5]]))
```

# Matrix Topic 9: Matrix-Vector & Matrix-Matrix Norms / Broadcasting-Style Ops

### 1. Idea (Definition)
Building on scalar norms and axis reductions, this topic covers two foundational matrix-and-vector operations used across numerical linear algebra and machine learning:

1. **Matrix-Vector Residual / Norm-Based Distance (`mat_vec_residual_norm` or `mat_vec_diff_norm`):** Given a validated matrix $\mathbf{A}$ of shape $(m, n)$, a vector $\mathbf{x}$ of length $n$, and a target vector $\mathbf{b}$ of length $m$, computes the Euclidean ($L_2$) norm of the linear residual vector $\mathbf{r} = \mathbf{A}\mathbf{x} - \mathbf{b}$:

   $$\|\mathbf{A}\mathbf{x} - \mathbf{b}\|_2 = \sqrt{\sum_{i=0}^{m-1} \left( (\mathbf{A}\mathbf{x})_i - b_i \right)^2}$$

2. **Row-Wise Vector Addition (`mat_add_row_vec`):** Given a validated matrix $\mathbf{A}$ of shape $(m, n)$ and a 1D vector $\mathbf{v}$ of length $n$, returns a new $(m, n)$ matrix where $\mathbf{v}$ is added to **every row** of $\mathbf{A}$ (manual NumPy-style row broadcasting):

   $$B_{i,j} = A_{i,j} + v_j \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

> **Note:** Dimension alignment is strict here—for $\mathbf{A}\mathbf{x} - \mathbf{b}$, $\mathbf{A}$ is $(m, n)$, $\mathbf{x}$ must have length $n$, and $\mathbf{b}$ must have length $m$. For row-wise vector addition, $\mathbf{v}$ must match the **column count** $n$ of $\mathbf{A}$, not the row count $m$.

---

### 2. Intuition
* **Residual norm intuition:** Measures how far $\mathbf{A}\mathbf{x}$ is from hitting target $\mathbf{b}$ in a linear system $\mathbf{A}\mathbf{x} \approx \mathbf{b}$. In least squares, optimization, and iterative solvers, shrinking $\|\mathbf{A}\mathbf{x} - \mathbf{b}\|_2$ toward zero measures fit quality and convergence.
* **Row-wise vector addition intuition:** Shifts every sample/row by the same offset vector—just like adding a **bias vector** $\mathbf{b}$ across a batch of activations or centering/shifting feature columns across rows.

---

### 3. Approach / Algorithm Behind It

* **Residual norm ($\|\mathbf{A}\mathbf{x} - \mathbf{b}\|_2$):** Validate the matrix and verify vector lengths match the required inner and outer dimensions. Combine your matrix-vector product logic with element-wise subtraction and $L_2$ norm aggregation, raising clear `ValueError`s on shape mismatches. Runs in $O(m \cdot n)$ time and $O(m)$ or $O(1)$ auxiliary space depending on whether you materialize the residual vector.
* **Row-wise vector addition:** Validate the matrix and check that the 1D vector length equals the number of columns $n$. Construct a new $(m, n)$ matrix adding each vector entry to the matching column across every row without mutating the input matrix. Runs in $O(m \cdot n)$ time and $O(m \cdot n)$ space.

```python
def mat_residual_norm(mat,x,b):
  m,n = mat_validate(mat)

  if not x or not b:
    raise ValueError("Vector and targeted vector cannot be empty.")
  if (n!= len(x)) or (m!=len(b)):
    raise ValueError("Dimension of matrix with vector or targeted vector mismatched.")

  r_sum =0

  for i in range(m):
    y=0
    for j in range(n):
      y+=(mat[i][j]*x[j])
    r_sum = r_sum + (y-b[i])**2

  return math.sqrt(r_sum)

print(mat_residual_norm([[1,2,3,4,5]],[7,4,1,8,5],[8]))
```

```python
def mat_add_row_vec(mat,vec):
  m,n = mat_validate(mat)
  if not vec:
    raise ValueError("Vector cannot be empty.")
  if n!= len(vec):
    raise ValueError("Dimension mismatched.")

  res = [[0 for _ in range(n)] for _ in range(m)]

  for i in range(m):
    for j in range(n):
      res[i][j]= mat[i][j]+vec[j]


  return res

print(mat_add_row_vec([[1,2,3,4,5]],[10,20,30,40,50]))
```

# Matrix Topic 10: Matrix-Matrix Distance & Column-Wise Vector Addition

### 1. Idea (Definition)
Continuing from row-wise broadcasting and norm-based error measures, this topic pairs a full-matrix norm difference with column-wise (row-broadcast) vector addition:

1. **Matrix-Matrix Frobenius Distance (`mat_frobenius_dist`):** Given two validated matrices $\mathbf{A}$ and $\mathbf{B}$ of the **exact same shape** $(m, n)$, computes the Frobenius norm of their difference matrix $(\mathbf{A} - \mathbf{B})$:

   $$\|\mathbf{A} - \mathbf{B}\|_F = \sqrt{\sum_{i=0}^{m-1} \sum_{j=0}^{n-1} \left( A_{i,j} - B_{i,j} \right)^2}$$

2. **Column-Wise Vector Addition (`mat_add_col_vec`):** Given a validated matrix $\mathbf{A}$ of shape $(m, n)$ and a 1D vector $\mathbf{v}$ of length $m$, returns a new $(m, n)$ matrix where entry $v_i$ is added to **every column in row $i$**:

   $$C_{i,j} = A_{i,j} + v_i \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

> **Note:** Watch the vector length carefully—for `mat_add_col_vec`, $\mathbf{v}$ must match the **row count** $m$ of $\mathbf{A}$ (not the column count $n$), because each element $v_i$ shifts an entire row $i$ across all $n$ columns.

---

### 2. Intuition
* **Frobenius distance intuition:** Treats two $(m, n)$ matrices as flat vectors of length $m \cdot n$ and measures their Euclidean distance apart. Useful for checking reconstruction error (e.g., low-rank approximations, autoencoders, matrix factorization) or comparing weight snapshots.
* **Column-wise vector addition intuition:** Applies a per-row offset across every feature/column in that row—such as adding a row-specific bias/intercept or scaling factor to each sample while keeping columns aligned.

---

### 3. Approach / Algorithm Behind It

* **Frobenius distance ($\|\mathbf{A} - \mathbf{B}\|_F$):** Confirm both matrices share the same validated dimensions, accumulate the squared entry-wise differences across all cells, and take the square root at the end. Runs in $O(m \cdot n)$ time with $O(1)$ auxiliary space.
* **Column-wise vector addition:** Validate the matrix and verify that the 1D vector length matches the number of rows $m$. Build a fresh $(m, n)$ matrix by adding each row's corresponding vector value across all columns without mutating the inputs. Runs in $O(m \cdot n)$ time and $O(m \cdot n)$ space.

```python
def mat_mat_frob(mat1,mat2):
  m1,n1 = mat_validate(mat1)
  m2,n2 = mat_validate(mat2)

  if (m1!=m2) or (n1!=n2):
    raise ValueError("Dimension of both matrices mismatched.")

  frob_sum =0
  for i in range(m1):
    for j in range(n1):
      frob_sum+=(mat1[i][j]-mat2[i][j])**2

  return math.sqrt(frob_sum)

print(mat_mat_frob([[1,4,7,8,5,2]],[[9,6,3,2,5,8]]))
```

```python
def mat_add_col_vec(mat,vec):
  m,n = mat_validate(mat)
  if not vec:
    raise ValueError("Vector cannot be empty.")
  if m!= len(vec):
    raise ValueError("Dimension mismatched.")

  res = [[0 for _ in range(n)] for _ in range(m)]

  for i in range(m):
    for j in range(n):
      res[i][j]= mat[i][j]+vec[i]


  return res

print(mat_add_col_vec([[1,2,3],[4,5,6]],[10,20]))
```

## Matrix Topic 11: Column-Wise Mean Centering

### 1. Idea (Definition)
Moving from reductions and vector broadcasting to structural matrix summaries and data preprocessing primitives, this topic covers two core operations:

1. **Column-Wise Mean Centering (`mat_center_cols`):** Given a validated matrix $\mathbf{A}$ of shape $(m, n)$, computes the mean of each column across all $m$ rows and subtracts each column's mean from every entry in that column, returning a new $(m, n)$ matrix:

   $$\tilde{A}_{i,j} = A_{i,j} - \mu_j, \quad \text{where } \mu_j = \frac{1}{m} \sum_{k=0}^{m-1} A_{k,j}$$

---

### 2. Intuition
* **Column-wise mean centering intuition:** Shifts every feature column so its average across all rows becomes $0$. Standardizing/centering columns is one of the first steps in PCA, covariance estimation, and linear regression before scaling or decomposing data.

---

### 3. Approach / Algorithm Behind It

* **Column-wise mean centering:** Validate the matrix (`m` rows, `n` columns). First compute the length-`n` column means vector `mu` (summing down each column `j` across `m` rows and dividing by `m`), then allocate a fresh `(m, n)` result matrix and set `res[i][j] = mat[i][j] - mu[j]` without mutating the input matrix. Runs in $O(m \cdot n)$ time and $O(m \cdot n)$ space (or $O(n)$ for the means vector plus the output matrix).

```python
def mat_mean_centering(mat):
  m,n=mat_validate(mat)
  mean_vec = [0 for _ in range(n)]
  for i in range(n):
    mean_col=0
    for j in range (m):
      mean_col += mat[j][i]
    mean_vec[i]=mean_col/m

  res =[[0 for _ in range(n)] for _ in range(m)]

  for i in range(m):
    for j in range(n):
      res[i][j]= mat[i][j] - mean_vec[j]

  return res

print(mat_mean_centering([[1,2,3,4],[5,6,7,8]]))
```

# Matrix Topic 12: Column-Wise & Row-Wise L2 Normalization

### 1. Idea (Definition)
Normalization scales vectors along a chosen axis so that each vector has a unit Euclidean ($L_2$) length of $1$. Given a validated matrix $\mathbf{A}$ of shape $(m, n)$:

1. **Column-Wise L2 Normalization (`mat_norm_cols`):** Computes the $L_2$ norm of each column across all $m$ rows and scales every entry in that column by its norm, returning a new $(m, n)$ matrix:

   $$\hat{A}_{i,j} = \frac{A_{i,j}}{\|\mathbf{A}_{:,j}\|_2}, \quad \text{where } \|\mathbf{A}_{:,j}\|_2 = \sqrt{\sum_{k=0}^{m-1} A_{k,j}^2}$$

2. **Row-Wise L2 Normalization (`mat_norm_rows`):** Computes the $L_2$ norm of each row across all $n$ columns and scales every entry in that row by its norm, returning a new $(m, n)$ matrix:

   $$\hat{A}_{i,j} = \frac{A_{i,j}}{\|\mathbf{A}_{i,:}\|_2}, \quad \text{where } \|\mathbf{A}_{i,:}\|_2 = \sqrt{\sum_{k=0}^{n-1} A_{i,k}^2}$$

> **Note:** Watch out for **zero-norm vectors** (an all-zero column or all-zero row). Decide how your library should handle division by zero—either raising a `ValueError` or keeping the zero vector unchanged.

---

### 2. Intuition
* **Column-wise normalization:** Puts every feature column on the same unit scale ($\|\mathbf{col}_j\|_2 = 1$), preventing large-magnitude features from dominating dot products, gradients, or distances.
* **Row-wise normalization:** Projects every sample/row onto the unit sphere ($\|\mathbf{row}_i\|_2 = 1$). Once rows are unit-normalized, the dot product between any two rows equals their **cosine similarity** directly.

---

### 3. Approach / Algorithm Behind It
* **Column-wise normalization:** Validate the matrix first. Find the Euclidean length of each column independently, handle any zero-norm columns according to your contract, and produce a new $(m, n)$ matrix with each column scaled down by its own norm without mutating the input. Runs in $O(m \cdot n)$ time and $O(m \cdot n)$ output space.
* **Row-wise normalization:** Validate the matrix first. Since each row is self-contained, you can compute a row's Euclidean length, guard against zero norms, and scale that row directly into the new $(m, n)$ output matrix. Runs in $O(m \cdot n)$ time and $O(m \cdot n)$ output space.

```python
def mat_norm_cols(mat, allow_zero=False):
  m, n = mat_validate(mat)
  col_norms = [0.0 for _ in range(n)]

  for j in range(n):
    sq_sum = 0.0
    for i in range(m):
      sq_sum += mat[i][j] ** 2
    norm = math.sqrt(sq_sum)
    if norm == 0.0 and not allow_zero:
      raise ValueError(f"Column {j} has zero norm; cannot normalize.")
    col_norms[j] = norm

  res = [[0.0 for _ in range(n)] for _ in range(m)]
  for i in range(m):
    for j in range(n):
      if col_norms[j] != 0.0:
        res[i][j] = mat[i][j] / col_norms[j]

  return res

print("Col norm:", mat_norm_cols([[3, 0], [4, 0]], allow_zero=True))
```

```python
def mat_norm_rows(mat, allow_zero=False):
  m, n = mat_validate(mat)
  res = [[0.0 for _ in range(n)] for _ in range(m)]

  for i in range(m):
    sq_sum = 0.0
    for j in range(n):
      sq_sum += mat[i][j] ** 2
    row_norm = math.sqrt(sq_sum)
    if row_norm == 0.0:
      if not allow_zero:
        raise ValueError(f"Row {i} has zero norm; cannot normalize.")
      continue
    for j in range(n):
      res[i][j] = mat[i][j] / row_norm

  return res

print("Row norm:", mat_norm_rows([[3, 4], [0, 0]], allow_zero=True))
```

# Matrix Topic 13: Gram-Style Matrix Products ($\mathbf{A}^\top \mathbf{A}$ and $\mathbf{A}\mathbf{A}^\top$)

### 1. Idea (Definition)
Given a validated matrix $\mathbf{A}$ of shape $(m, n)$ ($m$ rows and $n$ columns), a Gram-style product multiplies $\mathbf{A}$ by its own transpose $\mathbf{A}^\top$:

1. **Column-Gram Matrix ($\mathbf{A}^\top \mathbf{A}$):** Has shape $(n, n)$. Entry $(j, k)$ is the dot product of column $j$ and column $k$ of $\mathbf{A}$:

   $$(\mathbf{A}^\top \mathbf{A})_{j,k} = \sum_{i=0}^{m-1} A_{i,j}\, A_{i,k} \quad \text{for } 0 \le j, k < n$$

2. **Row-Gram Matrix ($\mathbf{A}\mathbf{A}^\top$):** Has shape $(m, m)$. Entry $(i, k)$ is the dot product of row $i$ and row $k$ of $\mathbf{A}$:

   $$(\mathbf{A}\mathbf{A}^\top)_{i,k} = \sum_{j=0}^{n-1} A_{i,j}\, A_{k,j} \quad \text{for } 0 \le i, k < m$$

> **Note:** Both $\mathbf{A}^\top \mathbf{A}$ and $\mathbf{A}\mathbf{A}^\top$ are always **square** and **symmetric** for *any* rectangular $(m, n)$ matrix—even when $m \neq n$—because the dot product of two vectors is symmetric ($\mathbf{u} \cdot \mathbf{v} = \mathbf{v} \cdot \mathbf{u}$).

---

### 2. Intuition
* **$\mathbf{A}^\top \mathbf{A}$ intuition:** Measures pairwise inner products across **columns** (features), producing an $(n \times n)$ matrix. When columns are mean-centered, $\frac{1}{m-1}\mathbf{A}^\top\mathbf{A}$ is the sample covariance matrix, and $\mathbf{A}^\top\mathbf{A}$ appears directly in least-squares normal equations.
* **$\mathbf{A}\mathbf{A}^\top$ intuition:** Measures pairwise inner products across **rows** (samples/observations), producing an $(m \times m)$ matrix. In machine learning and kernel methods, this is the linear kernel (Gram) matrix capturing similarity between every pair of samples.

---

### 3. Approach / Algorithm Behind It
* **Validate first:** Always validate `mat` (`m` rows, `n` columns) using your existing contract before computing either product.
* **Choose your computation path:** You can either reuse your existing `mat_transpose` and `mat_mat_mul` building blocks or write dedicated inner-product loops over rows/columns—and optionally take advantage of symmetry so off-diagonal pairs aren't computed twice.
* **Complexity:**
  * $\mathbf{A}^\top \mathbf{A}$: $O(m \cdot n^2)$ time and $O(n^2)$ output space.
  * $\mathbf{A}\mathbf{A}^\top$: $O(m^2 \cdot n)$ time and $O(m^2)$ output space.

```python
def mat_gram_style(mat,mode='row'):
  if mode =='row':
    prod_mat = mat_mat_mul(mat,mat_transpose(mat))
  elif mode =='col':
    prod_mat = mat_mat_mul(mat_transpose(mat),mat)
  else:
    raise ValueError("You need to define mode.")

  return prod_mat

print(mat_gram_style([[1,2,3,4,5]],mode='row'))
```
