## Matrix Topic 1: Matrix Addition & Subtraction

### 1. Idea (Definition)
Matrix addition ($\mathbf{A} + \mathbf{B}$) and subtraction ($\mathbf{A} - \mathbf{B}$) take **two 2D matrices of the exact same shape** $(m, n)$ ($m$ rows and $n$ columns) and produce a **new 2D matrix of shape $(m, n)$** by adding or subtracting corresponding entries at each row $i$ and column $j$:

$$(\mathbf{A} \pm \mathbf{B})_{i,j} = A_{i,j} \pm B_{i,j} \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

> **Note:** Unlike the outer product, element-wise matrix addition and subtraction **require both matrices to have identical dimensions** ($m_A = m_B$ and $n_A = n_B$). Also, in pure Python list-of-lists form, a "jagged" input (where rows have different lengths) isn't a valid rectangular matrix—always validate shape consistency!

---

### 2. Intuition
* **Grid overlay intuition:** Think of two spreadsheets or pixel grids of the exact same width and height stacked on top of each other. Every cell $(i, j)$ combines only with the cell at the exact same position $(i, j)$—no row/column mixing happens here.
* **ML / Feature intuition:** In machine learning, element-wise matrix addition and subtraction show up in **residual / skip connections** ($\mathbf{x} + F(\mathbf{x})$ in ResNets and Transformers), **gradient updates** ($\mathbf{W} - \eta \nabla \mathbf{W}$), and **computing differences across batches or feature tables**.

---

### 3. Approach / Algorithm Behind It
1. **Validate non-empty & rectangular shape:**
   * Check that `A` and `B` are non-empty (`len(A) > 0` and `len(B) > 0`), and every row in `A` and `B` is non-empty and has a consistent row length (`len(row) == n`).
2. **Validate matching shapes:**
   * Ensure `m_A == m_B` (same number of rows) and `n_A == n_B` (same number of columns). Raise a `ValueError` if shapes differ.
3. **Allocate an $(m \times n)$ result grid:**
   * Build `res = [[0 for _ in range(n)] for _ in range(m)]` so each row is its own independent list.
4. **Nested iteration over rows and columns:**
   * Outer loop: `i` from `0` to `m - 1`.
   * Inner loop: `j` from `0` to `n - 1`, setting `res[i][j] = A[i][j] + B[i][j]` (or `-` for subtraction).
5. **Return the resulting 2D matrix:** Return `res`.

* **Time Complexity:** $O(m \cdot n)$, visiting every entry of the $m \times n$ grid once.
* **Space Complexity:** $O(m \cdot n)$ to store the resulting $m \times n$ matrix.

```python
def mat_shape_helper(mat1,mat2):
  if (not mat1) or (not mat2):
    raise ValueError("Matrix cannot be empty.")

  for i, row in enumerate(mat1):
    if not row:
      raise ValueError(f"Row {i} in matrix 1 is empty")

  for i, row in enumerate(mat2):
    if not row:
      raise ValueError(f"Row {i} in matrix 2 is empty")

  m1, n1 = len(mat1), len(mat1[0])
  m2, n2 = len(mat2), len(mat2[0])

  if any(len(row) != n1 for row in mat1):
    raise ValueError("Matrix 1 is jagged (rows have inconsistent lengths).")
  if any(len(row) != n2 for row in mat2):
    raise ValueError("Matrix 2 is jagged (rows have inconsistent lengths).")

  if m1 != m2 or n1 != n2:
    raise ValueError(
        f"Shape mismatch: ({m1}, {n1}) cannot align with ({m2}, {n2})."
    )
  return m1,n1
```

```python
def mat_add(mat1,mat2):
  m1,n1=mat_shape_helper(mat1,mat2)
  res = [[0 for _ in range(n1)] for _ in range(m1)]

  for i in range(m1):
    for j in range(n1):
      res[i][j]=mat1[i][j]+mat2[i][j]

  return res

print(mat_add([[1,2,3]],[[4,5,6]]))
```

```python
def mat_sub(mat1,mat2):
  m1,n1=mat_shape_helper(mat1,mat2)
  res = [[0 for _ in range(n1)] for _ in range(m1)]

  for i in range(m1):
    for j in range(n1):
      res[i][j]=mat1[i][j]-mat2[i][j]

  return res

print(mat_sub([[1,2,3]],[[4,5,6]]))
```

## Matrix Topic 2: Scalar Multiplication & Transpose

### 1. Idea (Definition)
This topic covers two foundational single-matrix operations on a 2D matrix $\mathbf{A}$ of shape $(m, n)$ ($m$ rows and $n$ columns):

1. **Scalar Multiplication ($\alpha \mathbf{A}$):** Multiplies every single entry of $\mathbf{A}$ by a scalar $\alpha$, producing a **new matrix of the same shape $(m, n)$**:

   $$(\alpha \mathbf{A})_{i,j} = \alpha \cdot A_{i,j} \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

2. **Matrix Transpose ($\mathbf{A}^\top$):** Flips $\mathbf{A}$ across its main diagonal by turning row $i$ into column $i$ and column $j$ into row $j$, producing a **new matrix of shape $(n, m)$**:

   $$(\mathbf{A}^\top)_{j,i} = A_{i,j} \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

> **Note:** Transpose changes shape from $(m, n)$ to $(n, m)$—a $(2, 3)$ matrix becomes $(3, 2)$. Also, always build a **new** list-of-lists rather than mutating input rows in place, especially when $m \neq n$. And just like before, validate that the matrix is non-empty with non-empty, equal-length (non-jagged) rows!

---

### 2. Intuition
* **Scalar Multiplication intuition:** Scales every entry in the grid uniformly—like turning up/down brightness on an image or scaling all weights/learning-rate updates ($\eta \nabla \mathbf{W}$) by a constant factor while keeping the layout $(m, n)$ intact.
* **Transpose intuition:** Rotates/reflects a table so rows become columns and columns become rows. In ML and linear algebra, transpose aligns shapes for multiplication ($\mathbf{X}^\top \mathbf{W}$, $\mathbf{Q}\mathbf{K}^\top$ in attention) or converts row vectors into column vectors and vice versa.

---

### 3. Approach / Algorithm Behind It

#### A. Single-Matrix Validation Helper (`_validate_matrix(mat)`)
1. Check `if not mat:` $\rightarrow$ raise `ValueError("Matrix cannot be empty.")`.
2. Check every row is non-empty (`if not row:` $\rightarrow$ raise `ValueError(...)`).
3. Set `rows = len(mat)` and `cols = len(mat[0])`; check `any(len(row) != cols for row in mat)` $\rightarrow$ raise `ValueError("Matrix is jagged.")`.
4. Return `rows, cols`.

#### B. Scalar Multiplication (`mat_scalar(mat, scalar)`)
1. Validate via `_validate_matrix(mat)` to get `rows, cols`.
2. Allocate `res = [[0 for _ in range(cols)] for _ in range(rows)]`.
3. Loop `i` in `range(rows)` and `j` in `range(cols)`: set `res[i][j] = scalar * mat[i][j]`.
4. Return `res`.

* **Time Complexity:** $O(m \cdot n)$ | **Space Complexity:** $O(m \cdot n)$

#### C. Matrix Transpose (`mat_transpose(mat)`)
1. Validate via `_validate_matrix(mat)` to get `rows, cols` (shape $(m, n)$).
2. Allocate `res = [[0 for _ in range(rows)] for _ in range(cols)]`—notice `rows` inner elements and `cols` outer rows, giving shape $(n, m)$!
3. Loop `i` in `range(rows)` and `j` in `range(cols)`: set `res[j][i] = mat[i][j]`.
4. Return `res`.

* **Time Complexity:** $O(m \cdot n)$ | **Space Complexity:** $O(n \cdot m)$ (which equals $O(m \cdot n)$ total entries)

```python
def mat_validate(mat):
  if not mat:
    raise ValueError("Matrix cannot be empty.")

  for i, row in enumerate(mat):
    if not row:
      raise ValueError(f"Row {i} in matrix is empty")


  m,n = len(mat), len(mat[0])

  if any(len(row) != n for row in mat):
    raise ValueError("Matrix is jagged (rows have inconsistent lengths).")

  return m,n
```

```python
def mat_scalar_mul(mat,num):
  m,n = mat_validate(mat)
  res =[[0 for _ in range(n)] for _ in range(m)]

  for i in range(m):
    for j in range(n):
      res[i][j] = num*mat[i][j]
  return res

print(mat_scalar_mul([[9,8,7,6,5,4,3,2,1]],8))
```

```python
def mat_transpose(mat):
  m,n = mat_validate(mat)
  res =[[0 for _ in range(m)] for _ in range(n)]

  for i in range(m):
    for j in range(n):
      res[j][i] = mat[i][j]
  return res

print(mat_transpose([[8,5,2]]))
```

## Matrix Topic 3: Matrix-Vector Multiplication ($\mathbf{A}\mathbf{x}$)

### 1. Idea (Definition)
Matrix-vector multiplication takes a 2D matrix $\mathbf{A}$ of shape $(m, n)$ ($m$ rows and $n$ columns) and a 1D vector $\mathbf{x}$ of length $n$, and produces a **1D output vector $\mathbf{y}$ of length $m$**.

Each output component $y_i$ (for row index $i \in \{0, 1, \dots, m-1\}$) is the **dot product** of the $i$-th row of $\mathbf{A}$ with the vector $\mathbf{x}$:

$$y_i = (\mathbf{A}\mathbf{x})_i = \sum_{j=0}^{n-1} A_{i,j}\, x_j = A_{i,0}x_0 + A_{i,1}x_1 + \dots + A_{i,n-1}x_{n-1}$$

> **Note on Shape Compatibility:** The number of **columns** in $\mathbf{A}$ ($n$) must equal the **length** of vector $\mathbf{x}$ ($n$):
> $$(m, \mathbf{n}) \times (\mathbf{n},) \longrightarrow (m,)$$
> If `n != len(x)` or `x` is empty, raise a `ValueError`. Also notice the output is a **1D list of length $m$** (not an $m \times 1$ nested list).

---

### 2. Intuition
* **Linear combination of columns / bank of dot products:**
  * **Row-dot-product view:** Every row of $\mathbf{A}$ acts like a weight vector/filter, and $y_i$ measures how strongly row $i$ aligns with $\mathbf{x}$.
  * **Column linear-combination view:** $\mathbf{A}\mathbf{x} = x_0\,\mathbf{a}_{:,0} + x_1\,\mathbf{a}_{:,1} + \dots + x_{n-1}\,\mathbf{a}_{:,n-1}$—the entries of $\mathbf{x}$ act as weights scaling each column of $\mathbf{A}$ and summing them up.
* **ML / Feature intuition:** A single dense/linear layer forward pass (before adding bias) is literally $\mathbf{z} = \mathbf{W}\mathbf{x}$ (or $\mathbf{x}^\top\mathbf{W}$), mapping an $n$-dimensional input feature vector into an $m$-dimensional output representation.

---

### 3. Approach / Algorithm Behind It
1. **Validate matrix $\mathbf{A}$:** Call `m, n = mat_validate(mat)` so $\mathbf{A}$ is guaranteed non-empty, has non-empty rows, and has uniform column width `n`.
2. **Validate vector $\mathbf{x}$:**
   * Check `if not vec:` $\rightarrow$ raise `ValueError("Vector cannot be empty.")`.
   * Check `if len(vec) != n:` $\rightarrow$ raise `ValueError(f"Dimension mismatch: matrix has {n} columns, but vector has length {len(vec)}.")`.
3. **Allocate 1D result vector of length `m`:** Initialize `res = [0 for _ in range(m)]`.
4. **Compute dot product for each row:**
   * Outer loop: `i` from `0` to `m - 1`.
   * Initialize `row_sum = 0`.
   * Inner loop: `j` from `0` to `n - 1`, accumulating `row_sum += mat[i][j] * vec[j]`.
   * Store `res[i] = row_sum`.
5. **Return `res`:** Return the 1D list of length `m`.

* **Time Complexity:** $O(m \cdot n)$, performing $m \cdot n$ multiplications and additions across all rows and columns.
* **Space Complexity:** $O(m)$ for the output vector of length $m$ (plus $O(1)$ extra working variables).

```python
def mat_vec_mul(mat,vec):
  m,n = mat_validate(mat)
  if not vec:
    raise ValueError("Vector cannot be empty.")
  if len(vec)!=n:
    raise ValueError(f"Dimension mismatched: matrix has {n} columns, but vector has length {len(vec)}.")

  res = [0 for _ in range(m)]

  for i in range(m):
    row_sum =0
    for j in range(n):
      row_sum += (mat[i][j]*vec[j])
    res[i] = row_sum

  return res

print(mat_vec_mul([[1,2,3,4,5]],[1,4,7,2,5]))
```

## Matrix Topic 4: Matrix-Matrix Multiplication ($\mathbf{A}\mathbf{B}$)

### 1. Idea (Definition)
Matrix-matrix multiplication takes a 2D matrix $\mathbf{A}$ of shape $(m, k)$ ($m$ rows and $k$ columns) and a 2D matrix $\mathbf{B}$ of shape $(k, n)$ ($k$ rows and $n$ columns), and produces a **2D output matrix $\mathbf{C} = \mathbf{A}\mathbf{B}$ of shape $(m, n)$**.

Each entry $C_{i,j}$ (for row $i \in \{0, \dots, m-1\}$ and column $j \in \{0, \dots, n-1\}$) is the **dot product of the $i$-th row of $\mathbf{A}$ with the $j$-th column of $\mathbf{B}$**:

$$C_{i,j} = (\mathbf{A}\mathbf{B})_{i,j} = \sum_{p=0}^{k-1} A_{i,p}\, B_{p,j} = A_{i,0}B_{0,j} + A_{i,1}B_{1,j} + \dots + A_{i,k-1}B_{k-1,j}$$

> **Note on Shape Compatibility:** The inner dimensions must match—the number of **columns in $\mathbf{A}$** ($k$) must equal the number of **rows in $\mathbf{B}$** ($k$):
> $$(m, \mathbf{k}) \times (\mathbf{k}, n) \longrightarrow (m, n)$$
> If `n_A != m_B`, raise a `ValueError`. Also remember: matrix multiplication is **not commutative** in general ($\mathbf{A}\mathbf{B} \neq \mathbf{B}\mathbf{A}$), even when both products are valid!

---

### 2. Intuition
* **Grid of dot products:** Think of $\mathbf{C}$ as an $(m \times n)$ scorecard where cell $(i, j)$ answers: *"What is the dot product between row $i$ of $\mathbf{A}$ and column $j$ of $\mathbf{B}$?"*
* **Batching matrix-vector products:** Multiplying $\mathbf{A}$ $(m, k)$ by $\mathbf{B}$ $(k, n)$ is equivalent to multiplying $\mathbf{A}$ by each column of $\mathbf{B}$ one by one and placing those resulting column vectors side by side as the columns of $\mathbf{C}$.
* **ML / Feature intuition:** A mini-batch linear layer $\mathbf{Y} = \mathbf{X}\mathbf{W}$ (where $\mathbf{X}$ has shape `(batch_size, in_features)` and $\mathbf{W}$ has shape `(in_features, out_features)`) or scaled dot-product attention scores $\mathbf{Q}\mathbf{K}^\top$ is built directly on this operation.

---

### 3. Approach / Algorithm Behind It
1. **Validate both matrices:**
   * Call `m1, n1 = mat_validate(mat1)` and `m2, n2 = mat_validate(mat2)`.
   * Here, `m1` is rows of `A`, `n1` (which equals `k`) is columns of `A`, `m2` is rows of `B`, and `n2` is columns of `B`.
2. **Validate inner dimension alignment:**
   * Check `if n1 != m2:` $\rightarrow$ raise `ValueError(f"Dimension mismatch: mat1 has {n1} columns, but mat2 has {m2} rows.")`.
3. **Allocate an $(m_1 \times n_2)$ result matrix:**
   * Initialize `res = [[0 for _ in range(n2)] for _ in range(m1)]`.
4. **Triple-nested loop (`i`, `j`, `p`):**
   * Outer loop: `i` from `0` to `m1 - 1` (row of `mat1`).
   * Middle loop: `j` from `0` to `n2 - 1` (column of `mat2`).
   * Inner loop: `p` from `0` to `n1 - 1` (shared inner dimension `k`), accumulating:
     `dot_sum += mat1[i][p] * mat2[p][j]`
   * Assign `res[i][j] = dot_sum`.
5. **Return `res`:** Return the $(m_1 \times n_2)$ list of lists.

* **Time Complexity:** $O(m_1 \cdot n_1 \cdot n_2)$, performing $m_1 \cdot n_1 \cdot n_2$ scalar multiplications and additions.
* **Space Complexity:** $O(m_1 \cdot n_2)$ for the output matrix (plus $O(1)$ extra working variables).

```python
def mat_mat_mul(mat1,mat2):
  m1,n1 = mat_validate(mat1)
  m2,n2 = mat_validate(mat2)

  if n1 != m2 :
    raise ValueError(f"Dimension mismatch: mat1 has {n1} columns, but mat2 has {m2} rows.")

  res = [[0 for _ in range(n2)] for _ in range(m1)]

  for i in range(m1):
    for j in range(n2):
      dot_sum =0
      for p in range(n1):
        dot_sum +=(mat1[i][p]*mat2[p][j])
      res[i][j]=dot_sum
  return res

print(mat_mat_mul([[1,2,3]],[[1],[2],[3]]))
```

## Matrix Topic 5: Matrix Trace & Frobenius Norm

### 1. Idea (Definition)
This topic introduces two classic scalar summaries of a 2D matrix $\mathbf{A}$:

1. **Matrix Trace ($\operatorname{tr}(\mathbf{A})$):** Defined for a **square matrix** $\mathbf{A}$ of shape $(n, n)$, the trace is the **sum of the entries along the main diagonal** (where row index equals column index):

   $$\operatorname{tr}(\mathbf{A}) = \sum_{i=0}^{n-1} A_{i,i} = A_{0,0} + A_{1,1} + \dots + A_{n-1,n-1}$$

2. **Frobenius Norm ($\|\mathbf{A}\|_F$):** Defined for **any 2D matrix** $\mathbf{A}$ of shape $(m, n)$, the Frobenius norm is the **square root of the sum of squares of all its entries** (just like stretching all $m \cdot n$ entries into a single 1D vector and taking its $L_2$ norm):

   $$\|\mathbf{A}\|_F = \sqrt{\sum_{i=0}^{m-1} \sum_{j=0}^{n-1} A_{i,j}^2}$$

> **Note:** Trace requires a **square** matrix ($m == n$); if a matrix isn't square (e.g., shape `(2, 3)`), raise a `ValueError`. By contrast, the Frobenius norm works on **any valid rectangular** $(m, n)$ matrix—including non-square shapes.

---

### 2. Intuition
* **Trace intuition:** Summarizes the main diagonal of a square matrix in a single number. In linear algebra and ML, $\operatorname{tr}(\mathbf{A})$ equals the **sum of the eigenvalues** of $\mathbf{A}$, and shows up in **covariance matrix analysis** (total variance across features) and **matrix calculus** identities.
* **Frobenius norm intuition:** Measures the overall "size" or "energy" of a whole 2D table/matrix—equivalent to flattening the $(m \times n)$ grid into a vector of length $m \cdot n$ and computing its Euclidean ($L_2$) length. In ML, $\|\mathbf{W}\|_F^2$ appears in **weight decay / $L_2$ regularization**, and $\|\mathbf{A} - \hat{\mathbf{A}}\|_F$ measures **matrix reconstruction error**.

---

### 3. Approach / Algorithm Behind It

#### A. Matrix Trace (`mat_trace(mat)`)
1. Validate via `m, n = mat_validate(mat)`.
2. Check `if m != n:` $\rightarrow$ raise `ValueError(f"Trace requires a square matrix, got shape ({m}, {n}).")`.
3. Initialize `trace_sum = 0`.
4. Loop `i` from `0` to `n - 1`: accumulate `trace_sum += mat[i][i]`.
5. Return `trace_sum`.

* **Time Complexity:** $O(n)$ (only visits the $n$ diagonal elements of an $n \times n$ matrix).
* **Space Complexity:** $O(1)$ extra space.

#### B. Frobenius Norm (`mat_frobenius_norm(mat)`)
1. Validate via `m, n = mat_validate(mat)`.
2. Initialize `sq_sum = 0`.
3. Outer loop `i` from `0` to `m - 1`, inner loop `j` from `0` to `n - 1`: accumulate `sq_sum += mat[i][j] ** 2` (or `mat[i][j] * mat[i][j]`).
4. Return `math.sqrt(sq_sum)`.

* **Time Complexity:** $O(m \cdot n)$ (visits every entry in the $m \times n$ matrix).
* **Space Complexity:** $O(1)$ extra space.

```python
def mat_trace(mat):
  m,n = mat_validate(mat)
  if m!=n:
    raise ValueError(f"Trace requires a square matrix, got shape ({m},{n}).")

  trace =0
  for i in range(n):
    trace += mat[i][i]
  return trace

print(mat_trace([[1,20],[4,5]]))
```

```python
import math
def mat_frobenius_norm(mat):
  m,n = mat_validate(mat)

  frob_sum =0
  for i in range(m):
    for j in range(n):
      frob_sum += (mat[i][j]**2)

  return math.sqrt(frob_sum)

print(mat_frobenius_norm([[7,5,3,2,6,9,4,1]]))
```
