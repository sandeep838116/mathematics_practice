# Vector Addition

### 1. Idea (Definition)
Vector addition combines two vectors of the same dimension (same number of elements) to produce a new vector of that same dimension. Each component of the resulting vector is the sum of the corresponding components of the input vectors.
---
### 2. Approach / Algorithm Behind It :
- a. Validate inputs: Check that both vectors are non-empty and have equal length $n$.
- b. If their dimensions differ, the operation is undefined—flag a dimension mismatch error before doing any math.
- c. Allocate output space: Prepare a result container of size $n$ to hold the summed components.
- d. Traverse index by index: Loop through index $i$ from $0$ to $n-1$ (or $1$ to $n$ in mathematical notation), add the $i$-th element of the first vector to the $i$-th element of the second vector, and store the sum at index $i$ of the output.
- e. Return the resulting vector: Once all $n$ positions are processed, return the new vector.
- Time Complexity: $O(n)$, because each of the $n$ components is visited once.
- Space Complexity: $O(n)$ for a new output vector (or $O(1)$ auxiliary space if updating one vector in place).

```python
def vec_add(vec1, vec2):
    if len(vec1) != len(vec2):
        raise ValueError("Dimensions of the two vectors are different.")
    if not vec1:
        raise ValueError("Vectors cannot be empty.")

    n = len(vec1)
    res = [0] * n
    for i in range(n):
        res[i] = vec1[i] + vec2[i]

    return res
```

```python
print(vec_add([1,2,3,4,5],[6,7,8,9,5]))
```

```python
import numpy as np
def np_vec_add(vec1, vec2):
  a, b = np.asarray(vec1), np.asarray(vec2)
  if a.ndim != 1 or b.ndim != 1 or a.size == 0 or a.shape != b.shape:
    raise ValueError("Inputs must be non-empty 1D vectors of equal length.")
  return a + b

print(np_vec_add([1,2,3,4,5],[6,7,8,9,5]))
```

# Topic 2: Vector Subtraction

### 1. Idea (Definition)
Vector subtraction takes two vectors of the **same dimension** ($n$) and produces a new vector of dimension $n$ where each element is the difference between the corresponding components of the first vector and the second vector ($\mathbf{a} - \mathbf{b}$).

> **Note:** Order matters! Unlike vector addition, vector subtraction is **not commutative** ($\mathbf{a} - \mathbf{b} \neq \mathbf{b} - \mathbf{a}$ in general; instead, $\mathbf{b} - \mathbf{a} = -(\mathbf{a} - \mathbf{b})$).

---

### 2. Approach / Algorithm Behind It
1. **Validate dimensions:** Check that both vectors have equal length $n$. If `len(vec1) != len(vec2)`, stop and raise a dimension mismatch error.
2. **Validate non-empty inputs:** Ensure the vectors are non-empty (`n > 0`).
3. **Allocate output container:** Create a result list/array of size $n$.
4. **Iterate and compute difference:** Loop index $i$ from $0$ to $n - 1$, compute `vec1[i] - vec2[i]`, and store it at index $i$ of the result.
5. **Return the result:** Return the resulting difference vector.

* **Time Complexity:** $O(n)$, visiting each element once.
* **Space Complexity:** $O(n)$ for a new output vector (or $O(1)$ auxiliary space if mutating in place).

```python
def vec_sub(vec1,vec2):
  if len(vec1) != len(vec2):
    raise ValueError("Dimensions of both vector does not match.")
  if not vec1:
    raise ValueError("Vectors cann't be empty.")
  n = len(vec1)
  res = [0]*n
  for i in range(n):
    res[i] = vec1[i]-vec2[i]
  return res


print(vec_sub([9,8,7,6],[1,2,3,4]))
```

```python
def np_vec_sub(vec1, vec2):
  a, b = np.asarray(vec1), np.asarray(vec2)
  if a.ndim != 1 or b.ndim != 1 or a.size == 0 or a.shape != b.shape:
    raise ValueError("Inputs must be non-empty 1D vectors of equal length.")
  return a - b
print(np_vec_sub([9,8,7,6],[1,2,3,4]))
```

### Topic 3: Scalar Multiplication

### 1. Idea (Definition)
Scalar multiplication takes a single vector $\mathbf{a}$ of dimension $n$ and a single number (scalar) $c$, producing a new vector of dimension $n$ where every element of $\mathbf{a}$ is multiplied by $c$.


---

### 2. Intuition
* **Geometric intuition:** Multiplying a vector by a scalar stretches, shrinks, or flips its arrow without changing the line along which it points:
  * If $|c| > 1$, the vector stretches longer.
  * If $0 < |c| < 1$, the vector shrinks shorter.
  * If $c < 0$, the vector flips to the **exact opposite direction**.
  * If $c = 0$, every component becomes $0$ (the zero vector).
* **ML / Feature intuition:** Scaling a vector uniformly adjusts its overall strength or step size—such as multiplying a gradient vector by a **learning rate** ($\alpha \cdot \nabla L$) during gradient descent, or scaling a feature/embedding vector by a weight.

---

### 3. Approach / Algorithm Behind It
1. **Validate input vector:** Check that the vector is non-empty (`n > 0`). (If `n == 0`, decide whether to raise a `ValueError` or return `[]`—raising a `ValueError` keeps your convention consistent.)
2. **Allocate output space:** Create a result list/array of length $n$.
3. **Iterate and scale each component:** Loop index $i$ from $0$ to $n - 1$, compute `scalar * vec[i]`, and store it at `res[i]`.
4. **Return the scaled vector:** Return `res`.

* **Time Complexity:** $O(n)$, visiting each of the $n$ components once.
* **Space Complexity:** $O(n)$ for a new vector (or $O(1)$ auxiliary space if scaling in place).

```python
def vec_scalar(vec,num):
  if not vec:
    raise ValueError("Vector cannot be empty.")

  n = len(vec)
  res =[1]*n
  for i in range(n):
    res[i]=vec[i]*num
  return res

print(vec_scalar([1,2,3,4,5],9))
```

```python
def np_vec_scalar(vec, num):
  a = np.asarray(vec)
  if a.ndim != 1 or a.size == 0:
    raise ValueError("Input must be a non-empty 1D vector.")
  return a * num
print(np_vec_scalar([1,2,3,4,5],9))
```

# Topic 4: Dot Product (Inner Product)

### 1. Idea (Definition)
The dot product (also called the **inner product**) takes two vectors of the **same dimension** ($n$) and returns a **single scalar number** (not a vector). It is computed by multiplying corresponding components of the two vectors and summing all those products together:

$$\mathbf{a} \cdot \mathbf{b} = \sum_{i=0}^{n-1} a_i b_i = a_0 b_0 + a_1 b_1 + \dots + a_{n-1} b_{n-1}$$

> **Note:** Unlike vector addition or subtraction, the output is a **scalar**, not a vector of length $n$. Also, the dot product is **commutative** ($\mathbf{a} \cdot \mathbf{b} = \mathbf{b} \cdot \mathbf{a}$).

---

### 2. Intuition
* **Alignment & Direction intuition:** The dot product measures how much two vectors point in the **same general direction**, scaled by their lengths ($\mathbf{a} \cdot \mathbf{b} = \|\mathbf{a}\| \|\mathbf{b}\| \cos\theta$):
  * **Positive ($> 0$):** Angle between them is acute ($< 90^\circ$); they generally align.
  * **Zero ($= 0$):** Vectors are **orthogonal** (perpendicular, $90^\circ$ apart) or one is a zero vector—no alignment along each other.
  * **Negative ($< 0$):** Angle between them is obtuse ($> 90^\circ$); they point in opposing directions.
* **ML / Feature intuition:** In machine learning, the dot product is everywhere—computing a neuron's pre-activation ($\mathbf{w} \cdot \mathbf{x} + b$), scoring query-key similarity in Transformer **self-attention**, and measuring raw unnormalized similarity between embedding vectors.

---

### 3. Approach / Algorithm Behind It
1. **Validate dimensions:** Check that both vectors have equal length $n$. If `len(vec1) != len(vec2)`, raise a dimension mismatch error.
2. **Validate non-empty inputs:** Ensure `n > 0` (if empty, raise a `ValueError`).
3. **Initialize an accumulator:** Set a running sum variable (e.g., `dot_sum = 0` or `0.0`) before the loop.
4. **Iterate and accumulate products:** Loop index $i$ from $0$ to $n - 1$, multiply `vec1[i] * vec2[i]`, and add that product to `dot_sum`.
5. **Return the scalar total:** Return `dot_sum` after the loop finishes.

* **Time Complexity:** $O(n)$, visiting each pair of elements once.
* **Space Complexity:** $O(1)$ auxiliary space, since you only accumulate a single scalar total rather than allocating a new array.

```python
def vec_dot(vec1,vec2):
  if len(vec1)!=len(vec2):
    raise ValueError("Dimensions of the two vectors doesnot match.")
  if not vec1:
    raise ValueError("Vectors cannot be empty.")

  n = len(vec1)
  dot_sum =0
  for i in range(n):
    dot_sum+=(vec1[i]*vec2[i])

  return dot_sum

print(vec_dot([1,2,3,4,5],[7,8,9,4,6]))
```

```python
def np_vec_dot(vec1, vec2):
  a, b = np.asarray(vec1), np.asarray(vec2)
  if a.ndim != 1 or b.ndim != 1 or a.size == 0 or a.shape != b.shape:
    raise ValueError("Inputs must be non-empty 1D vectors of equal length.")
  return float(np.dot(a, b))

print(np_vec_dot([1,2,3,4,5],[7,8,9,4,6]))
```

# Topic 5: Cross Product

### 1. Idea (Definition)
The cross product ($\mathbf{a} \times \mathbf{b}$) is a binary operation defined specifically on **two 3-dimensional vectors** ($\mathbf{a} = [a_x, a_y, a_z]$ and $\mathbf{b} = [b_x, b_y, b_z]$). Unlike the dot product (which returns a scalar), the cross product returns a **new 3D vector** that is perpendicular (orthogonal) to both input vectors:

$$\mathbf{a} \times \mathbf{b} = \begin{bmatrix} a_y b_z - a_z b_y \\ a_z b_x - a_x b_z \\ a_x b_y - a_y b_x \end{bmatrix}$$

> **Note:** The cross product is **anti-commutative** (skew-symmetric): $\mathbf{b} \times \mathbf{a} = -(\mathbf{a} \times \mathbf{b})$. Flipping the order flips the direction of the resulting vector. Also, unlike the dot product, it is **not associative** in general ($\mathbf{a} \times (\mathbf{b} \times \mathbf{c}) \neq (\mathbf{a} \times \mathbf{b}) \times \mathbf{c}$).

---

### 2. Intuition
* **Geometric intuition:** Imagine two arrows lying flat on a table—that defines a plane. The cross product points straight **up or down out of that plane** (at $90^\circ$ to both arrows), following the **right-hand rule**. Its length equals $\|\mathbf{a}\|\|\mathbf{b}\|\sin\theta$, which is the **area of the parallelogram** formed by $\mathbf{a}$ and $\mathbf{b}$. If the two vectors are parallel or anti-parallel ($\theta = 0^\circ$ or $180^\circ$), $\sin\theta = 0$, so the cross product is the zero vector $[0, 0, 0]$.
* **ML / Engineering intuition:** While high-dimensional ML mostly uses dot products and matrix multiplications rather than 3D cross products, cross products are heavily used in **3D computer graphics, robotics, physics simulations, and spatial geometry**—for instance, computing surface normals of 3D triangles, checking orientation/winding order, or finding torque ($\boldsymbol{\tau} = \mathbf{r} \times \mathbf{F}$).

---

### 3. Approach / Algorithm Behind It
1. **Validate dimensions strictly:** Check that both vectors have length **exactly `3`** (`len(vec1) == 3` and `len(vec2) == 3`). If either is empty or has any dimension other than 3, raise a `ValueError`.
2. **Unpack or index the three components:** Let the components be $(a_0, a_1, a_2)$ for `vec1` and $(b_0, b_1, b_2)$ for `vec2`.
3. **Compute the three cross-product terms:**
   * `x = a_1 * b_2 - a_2 * b_1`
   * `y = a_2 * b_0 - a_0 * b_2`
   * `z = a_0 * b_1 - a_1 * b_0`
4. **Return the 3D result vector:** Return `[x, y, z]`.

* **Time Complexity:** $O(1)$, since the dimension is fixed at $3$ (a constant number of multiplications and subtractions).
* **Space Complexity:** $O(1)$ auxiliary space, returning a fixed 3-element vector.

```python
def vec_cross(vec1,vec2):
  if (len(vec1)!=3) or (len(vec2)!=3):
    raise ValueError("Dimension of the vectors must be 3.")
  if not vec1:
    raise ValueError("Vectors cannot be empty.")

  a_1,a_2,a_3 =vec1
  b_1,b_2,b_3 =vec2

  z = (a_1*b_2)-(a_2*b_1)
  y=(a_3*b_1)-(a_1*b_3)
  x = (a_2*b_3)-(a_3*b_2)
  return [x,y,z]

print(vec_cross([1,5,9],[7,5,3]))
```

```python
def np_vec_cross(vec1, vec2):
  a, b = np.asarray(vec1), np.asarray(vec2)
  if a.shape != (3,) or b.shape != (3,):
    raise ValueError("Both inputs must be 3D vectors of shape (3,).")
  return np.cross(a, b)
print(np_vec_cross([1,5,9],[7,5,3]))
```

# Topic 6: Hadamard Product (Element-wise Product)

### 1. Idea (Definition)
The Hadamard product ($\mathbf{a} \odot \mathbf{b}$) takes two vectors of the **same dimension** ($n$) and produces a **new vector of the same dimension $n$** where each element is the product of the corresponding elements of the two input vectors:

$$(\mathbf{a} \odot \mathbf{b})_i = a_i b_i \quad \text{for } i = 0, 1, \dots, n-1$$

> **Note:** Unlike the **dot product** (which multiplies corresponding elements and sums them into a single scalar), the Hadamard product **keeps each product separate** inside a vector of length $n$. Like vector addition and the dot product, it is **commutative** ($\mathbf{a} \odot \mathbf{b} = \mathbf{b} \odot \mathbf{a}$).

---

### 2. Intuition
* **Component-wise modulation intuition:** Think of one vector as a signal and the second vector as a **filter, mask, or gain control** applied to each dimension independently. Where the mask/weight is `0`, that component gets zeroed out; where it is `> 1`, that component is amplified; where it is negative, that component flips sign.
* **ML / Feature intuition:** In neural networks and ML, element-wise multiplication is heavily used for **gating mechanisms** (such as forget/input/output gates in LSTMs and GRUs, or SwiGLU/GLU activations in modern LLMs), **feature weighting/attention masks**, and **gradient scaling** during backpropagation.

---

### 3. Approach / Algorithm Behind It
1. **Validate dimensions:** Check that `len(vec1) == len(vec2)`. If they differ, raise a `ValueError` for dimension mismatch.
2. **Validate non-empty inputs:** Ensure `n > 0` (raise a `ValueError` if empty).
3. **Allocate output space:** Create a result list/array of size $n$.
4. **Iterate and multiply element-wise:** Loop index $i$ from $0$ to $n - 1$, compute `vec1[i] * vec2[i]`, and store it at `res[i]`.
5. **Return the resulting vector:** Return `res`.

* **Time Complexity:** $O(n)$, visiting each of the $n$ index positions once.
* **Space Complexity:** $O(n)$ for the output vector of length $n$.

```python
def vec_hadamard(vec1,vec2):
  if len(vec1)!=len(vec2):
    raise ValueError("Dimension of the vectors are not same.")
  if not vec1:
    raise ValueError("Vectors cannot be empty.")
  n=len(vec1)
  res=[1]*n
  for i in range(n):
    res[i]= vec1[i]*vec2[i]

  return res

print(vec_hadamard([7,4,1],[8,5,2]))
```

```python
def np_vec_hadamard(vec1, vec2):
  a, b = np.asarray(vec1), np.asarray(vec2)
  if a.ndim != 1 or b.ndim != 1 or a.size == 0 or a.shape != b.shape:
    raise ValueError("Inputs must be non-empty 1D vectors of equal length.")
  return a * b  # or np.multiply(a, b)

print(np_vec_hadamard([7,4,1],[8,5,2]))
```

# Topic 7: Outer Product

### 1. Idea (Definition)
The outer product ($\mathbf{a} \otimes \mathbf{b}$ or $\mathbf{a}\mathbf{b}^T$) takes **two 1D vectors**—say $\mathbf{a}$ of length $m$ and $\mathbf{b}$ of length $n$—and produces a **2D matrix of shape $(m, n)$** ($m$ rows and $n$ columns). Each entry at row $i$ and column $j$ is the product of the $i$-th element of $\mathbf{a}$ and the $j$-th element of $\mathbf{b}$:

$$(\mathbf{a} \otimes \mathbf{b})_{i, j} = a_i \cdot b_j \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

> **Note:** Unlike vector addition, subtraction, dot product, or Hadamard product, the outer product **does NOT require $\mathbf{a}$ and $\mathbf{b}$ to have the same length** ($m$ and $n$ can be different!), and its result is a **2D grid (matrix)** rather than a 1D vector or scalar. Also, $\mathbf{a} \otimes \mathbf{b}$ is the transpose of $\mathbf{b} \otimes \mathbf{a}$ (not equal in general).

---

### 2. Intuition
* **Pairwise interaction grid intuition:** Think of row $i$ as scaling the entire vector $\mathbf{b}$ by $a_i$ (or equivalently, column $j$ as scaling vector $\mathbf{a}$ by $b_j$). Every cell $(i, j)$ captures the pairwise interaction between feature/coordinate $i$ of the first vector and feature/coordinate $j$ of the second vector.
* **ML / Feature intuition:** In machine learning, outer products build **rank-1 matrices**, covariance/gram-style outer terms, and **bilinear / cross-feature interactions** (such as combining user features and item features into an interaction matrix, or forming weight updates from activation and error vectors).

---

### 3. Approach / Algorithm Behind It
1. **Validate non-empty inputs:** Check that both `vec1` (length $m$) and `vec2` (length $n$) are non-empty (`m > 0` and `n > 0`). If either is empty, raise a `ValueError`.
2. **Determine matrix dimensions:** Let `m = len(vec1)` (number of rows) and `n = len(vec2)` (number of columns).
3. **Allocate a 2D result grid:** Create an $m \times n$ matrix (a list of $m$ rows, where each row has $n$ elements—be careful to create a fresh row per `i` rather than sharing the same row reference across all rows!).
4. **Nested iteration over rows and columns:**
   * Outer loop: `i` from `0` to `m - 1` (selecting `vec1[i]`).
   * Inner loop: `j` from `0` to `n - 1` (selecting `vec2[j]`), computing `vec1[i] * vec2[j]`, and storing it at position `[i][j]`.
5. **Return the 2D matrix:** Return the resulting $m \times n$ list of lists.

* **Time Complexity:** $O(m \cdot n)$, because every pair $(i, j)$ out of $m \times n$ entries is computed once.
* **Space Complexity:** $O(m \cdot n)$ to store the resulting $m \times n$ matrix.

```python
def vec_outer(vec1,vec2):
  if not vec1 or not vec2:
    raise ValueError("Vectors cannot be empty.")

  m = len(vec1)
  n = len(vec2)
  res = [[0 for _ in range(n)] for _ in range(m)]

  for i in range(m):
    for j in range(n):
      res[i][j]=vec1[i]*vec2[j]

  return res

print(vec_outer([1,2,3,4,5],[8,5,2]))
```

```python
def np_vec_outer(vec1, vec2):
  a, b = np.asarray(vec1), np.asarray(vec2)
  if a.ndim != 1 or b.ndim != 1 or a.size == 0 or b.size == 0:
    raise ValueError("Both inputs must be non-empty 1D vectors.")
  return np.outer(a, b)
print(np_vec_outer([1,2,3,4,5],[8,5,2]))
```

# Topic 8: $L_1$ Norm (Manhattan Norm)

### 1. Idea (Definition)
The $L_1$ norm (also called the **Manhattan norm** or **taxicab norm**) takes a single vector $\mathbf{a}$ of dimension $n$ and returns a **non-negative scalar** equal to the sum of the **absolute values** of all its components:

$$\|\mathbf{a}\|_1 = \sum_{i=0}^{n-1} |a_i| = |a_0| + |a_1| + \dots + |a_{n-1}|$$

> **Note:** Because each term uses the **absolute value** $|a_i|$, negative components contribute positively to the norm instead of canceling out positive ones. Also, $\|\mathbf{a}\|_1 = 0$ if and only if $\mathbf{a}$ is the zero vector.

---

### 2. Intuition
* **Grid / Taxicab intuition:** Imagine navigating a city grid (like Manhattan) where you can only travel along horizontal and vertical streets—no diagonal shortcuts. Walking from the origin to a point $(x, y)$ takes $|x| + |y|$ blocks total. The $L_1$ norm measures that axis-aligned total distance.
* **ML / Feature intuition:** Because it sums absolute values linearly without squaring large differences, the $L_1$ norm treats outliers less aggressively than squared penalties and encourages **sparsity** when used as a regularization penalty (such as **Lasso** regression, pushing irrelevant feature weights all the way to `0`).

---

### 3. Approach / Algorithm Behind It
1. **Validate input vector:** Check that the vector is non-empty (`len(vec) > 0`). If empty, raise a `ValueError`.
2. **Initialize an accumulator:** Set a running sum variable (e.g., `l1_sum = 0` or `0.0`) before iterating.
3. **Iterate and sum absolute values:** Loop index `i` from `0` to `n - 1`, take the absolute value of `vec[i]`, and add it to `l1_sum`.
4. **Return the scalar total:** Return `l1_sum` after processing all $n$ elements.

* **Time Complexity:** $O(n)$, visiting each of the $n$ components once.
* **Space Complexity:** $O(1)$ auxiliary space, maintaining only a single scalar running sum.

```python
def vec_l1(vec):
  if not vec:
    raise ValueError("Vector cannot be empty.")
  n = len(vec)
  l1_sum = 0
  for i in range(n):
    l1_sum += abs(vec[i])
  return l1_sum
print(vec_l1([1,2,3,-7,8,9,-5,4,10]))
```

```python
def np_vec_l1(vec):
  a = np.asarray(vec)
  if a.ndim != 1 or a.size == 0:
    raise ValueError("Input must be a non-empty 1D vector.")
  return float(np.linalg.norm(a, ord=1))
print(np_vec_l1([1,2,3,-7,8,9,-5,4,10]))
```

# Topic 9: $L_2$ Norm (Euclidean Norm)

### 1. Idea (Definition)
The $L_2$ norm (also called the **Euclidean norm** or **vector magnitude / length**) takes a single vector $\mathbf{a}$ of dimension $n$ and returns a **non-negative scalar** equal to the square root of the sum of the **squares** of all its components:

$$\|\mathbf{a}\|_2 = \sqrt{\sum_{i=0}^{n-1} a_i^2} = \sqrt{a_0^2 + a_1^2 + \dots + a_{n-1}^2}$$

> **Note:** Squaring each component automatically makes every term non-negative (so negative values don't cancel out) and gives larger components more weight before taking the final square root. Also, $\|\mathbf{a}\|_2 = \sqrt{\mathbf{a} \cdot \mathbf{a}}$.

---

### 2. Intuition
* **Straight-line / crow's-flight intuition:** In 2D or 3D space, $\|\mathbf{a}\|_2$ is the **straight-line distance** from the origin to the tip of vector $\mathbf{a}$ (the Pythagorean theorem generalized to $n$ dimensions).
* **ML / Feature intuition:** In machine learning, the $L_2$ norm measures overall vector scale/energy—used in **Ridge regression / weight decay** (penalizing large weights smoothly so no single feature dominates), **Euclidean distance** ($\|\mathbf{a} - \mathbf{b}\|_2$), and **vector normalization** (dividing by $\|\mathbf{a}\|_2$ to get unit length).

---

### 3. Approach / Algorithm Behind It
1. **Validate input vector:** Check that the vector is non-empty (`len(vec) > 0`). If empty, raise a `ValueError`.
2. **Initialize a sum-of-squares accumulator:** Set `sq_sum = 0` (or `0.0`) before the loop.
3. **Iterate and accumulate squares:** Loop index `i` from `0` to `n - 1`, square `vec[i]` (`vec[i] * vec[i]` or `vec[i] ** 2`), and add it to `sq_sum`.
4. **Take the square root:** Compute $\sqrt{\text{sq\_sum}}$ (for example via `math.sqrt(sq_sum)` or `sq_sum ** 0.5`).
5. **Return the scalar magnitude:** Return the resulting norm value.

* **Time Complexity:** $O(n)$, visiting each of the $n$ components once (and taking a single constant-time square root at the end).
* **Space Complexity:** $O(1)$ auxiliary space, tracking only a single scalar accumulator.

```python
import math
def vec_l2(vec):
  if not vec:
    raise ValueError("Vector cannot be empty.")
  n = len(vec)
  l2_sum = 0
  for i in range(n):
    l2_sum += (vec[i]**2)

  return math.sqrt(l2_sum)

print(vec_l2([1,2,3,4,5]))
```

```python
def np_vec_l2(vec):
  a = np.asarray(vec, dtype=float)
  if a.ndim != 1 or a.size == 0:
    raise ValueError("Input must be a non-empty 1D vector.")
  return float(np.linalg.norm(a, ord=2))

print(np_vec_l2([1,2,3,4,5]))
```

# Topic 10: $L_\infty$ Norm (Max Norm)

### 1. Idea (Definition)
The $L_\infty$ norm (also called the **max norm**, **maximum norm**, or **Chebyshev norm**) takes a single vector $\mathbf{a}$ of dimension $n$ and returns a **non-negative scalar** equal to the **largest absolute value** among all its components:

$$\|\mathbf{a}\|_\infty = \max_{0 \le i < n} |a_i| = \max(|a_0|, |a_1|, \dots, |a_{n-1}|)$$

> **Note:** Always take the **absolute value** $|a_i|$ before comparing! A negative component like `-12` has magnitude `12`, so if the other components are `[3, -12, 8]`, the $L_\infty$ norm is `12`, not `8`. Also, $\|\mathbf{a}\|_\infty = 0$ if and only if every component is `0`.

---

### 2. Intuition
* **Worst-case / bottleneck intuition:** While $L_1$ sums every coordinate and $L_2$ blends all coordinates together via squares, the $L_\infty$ norm asks: *"What is the single biggest magnitude or worst-case deviation in any one dimension?"*
* **ML / Feature intuition:** In machine learning, the $L_\infty$ norm shows up in **adversarial robustness / adversarial attacks** (bounding the maximum perturbation allowed on any single pixel or feature, such as an $\epsilon$-ball in $L_\infty$), **gradient clipping / worst-case error bounds**, and **Chebyshev / grid distance** ($\|\mathbf{a} - \mathbf{b}\|_\infty$, where diagonal steps in a grid count as a single step along the dominant axis).

---

### 3. Approach / Algorithm Behind It
1. **Validate input vector:** Check that the vector is non-empty (`len(vec) > 0`). If empty, raise a `ValueError`—you need at least one element to have a valid maximum.
2. **Initialize the running maximum:** Set `max_val = abs(vec[0])` (using the first element's magnitude rather than `0` avoids any issues even if all values are negative or very small).
3. **Scan remaining components:** Loop index `i` from `1` to `n - 1` (or `0` to `n - 1`), compute `mag = abs(vec[i])`, and if `mag > max_val`, update `max_val = mag`.
4. **Return the maximum magnitude:** Return `max_val` after checking all $n$ elements.

* **Time Complexity:** $O(n)$, visiting each of the $n$ elements once to find the maximum magnitude.
* **Space Complexity:** $O(1)$ auxiliary space, keeping only a single scalar variable.

```python
def vec_max_norm(vec):
  if not vec:
    raise ValueError("Vector cannot be empty.")

  n = len(vec)
  max = abs(vec[0])
  for i in range(n):
    if abs(vec[i])> max:
      max = abs(vec[i])
  return max

print(vec_max_norm([-9,-8,15,-54,65,-89]))
```

```python
def np_vec_max_norm(vec):
  a = np.asarray(vec)
  if a.ndim != 1 or a.size == 0:
    raise ValueError("Input must be a non-empty 1D vector.")
  return float(np.linalg.norm(a, ord=np.inf))

print(np_vec_max_norm([-9,-8,15,-54,65,-89]))
```

# Topic 11: Vector Normalization (Unit Vector Conversion)

### 1. Idea (Definition)
Vector normalization (converting a vector into a **unit vector**, denoted $\hat{\mathbf{a}}$) takes a non-zero vector $\mathbf{a}$ of dimension $n$ and scales it by the reciprocal of its $L_2$ norm so that its new length (magnitude) is **equal to $1$**, while keeping its **exact same direction**:

$$\hat{\mathbf{a}} = \frac{\mathbf{a}}{\|\mathbf{a}\|_2} = \left[ \frac{a_0}{\|\mathbf{a}\|_2},\; \frac{a_1}{\|\mathbf{a}\|_2},\; \dots,\; \frac{a_{n-1}}{\|\mathbf{a}\|_2} \right]$$

> **Note:** A **zero vector** ($\mathbf{a} = [0, 0, \dots, 0]$) has $\|\mathbf{a}\|_2 = 0$ and **no defined direction**, so dividing by its norm is invalid (division by zero). Always guard against a zero norm!

---

### 2. Intuition
* **Direction-only intuition:** Think of a vector as having two parts: **direction** (which way it points) and **magnitude** (how long it is). Normalization strips away magnitude—shrinking or stretching the arrow until its tip sits right on the unit circle/sphere of radius $1$—leaving pure direction.
* **ML / Feature intuition:** In machine learning, unit-normalized embeddings put every vector on the **unit hypersphere** so differences in vector length (for example, a longer document or higher-norm embedding) don't skew comparisons. On unit vectors, **cosine similarity simplifies directly to the dot product** ($\hat{\mathbf{a}} \cdot \hat{\mathbf{b}} = \cos\theta$).

---

### 3. Approach / Algorithm Behind It
1. **Validate non-empty input:** Check that `vec` has length $n > 0$; if empty, raise a `ValueError`.
2. **Compute the $L_2$ norm:** Calculate the Euclidean length `norm = \|\mathbf{a}\|_2` across all $n$ components (reusing your $L_2$ norm logic).
3. **Check for zero norm:** If `norm == 0` (or `norm <= 1e-12` if handling floating-point near-zero), raise a `ValueError` because a zero vector cannot be normalized.
4. **Divide each component by `norm`:** Allocate a result list/array of size $n$, loop index `i` from `0` to `n - 1`, compute `vec[i] / norm`, and store it at `res[i]`.
5. **Return the unit vector:** Return `res` (whose own $L_2$ norm is $\approx 1.0$).

* **Time Complexity:** $O(n)$ (two linear passes of size $n$: one to compute the norm, one to divide each element).
* **Space Complexity:** $O(n)$ for the new normalized output vector (or $O(1)$ auxiliary space if scaling in place).

```python
import math
def vec_l2_eud(vec):
  n = len(vec)
  l2_sum = 0
  for i in range(n):
    l2_sum += (vec[i]**2)

  return math.sqrt(l2_sum)

def vec_normalize(vec):
  if not vec:
    raise ValueError("Vectors cannot be empty")

  norm = vec_l2_eud(vec)
  if (norm == 0) or (norm <=1e-12):
    raise ValueError("Zero vector cannot be normalized.")

  n = len(vec)
  normalize = [1]*n
  for i in range(n):
    normalize[i] = vec[i]/norm

  return normalize


print(vec_normalize([1,2,3,4,5]))
```

```python
def np_vec_normalize(vec):
  a = np.asarray(vec, dtype=float)
  if a.ndim != 1 or a.size == 0:
    raise ValueError("Input must be a non-empty 1D vector.")
  norm = np.linalg.norm(a, ord=2)
  if norm <= 1e-12:
    raise ValueError("Zero vector cannot be normalized.")
  return a / norm

print(np_vec_normalize([1,2,3,4,5]))
```

# Topic 12: Cosine Similarity

### 1. Idea (Definition)
Cosine similarity measures the **cosine of the angle $\theta$** between two non-zero vectors $\mathbf{a}$ and $\mathbf{b}$ of the same dimension $n$. It equals their dot product divided by the product of their $L_2$ norms (equivalently, the dot product of their unit-normalized vectors $\hat{\mathbf{a}}$ and $\hat{\mathbf{b}}$):

$$\text{cosine\_similarity}(\mathbf{a}, \mathbf{b}) = \cos\theta = \frac{\mathbf{a} \cdot \mathbf{b}}{\|\mathbf{a}\|_2 \, \|\mathbf{b}\|_2} = \frac{\sum_{i=0}^{n-1} a_i b_i}{\sqrt{\sum_{i=0}^{n-1} a_i^2} \, \sqrt{\sum_{i=0}^{n-1} b_i^2}}$$

> **Note:** Cosine similarity always lies in the range $[-1, 1]$:
> * **$+1$:** Same direction ($\theta = 0^\circ$)
> * **$0$:** Orthogonal / perpendicular ($\theta = 90^\circ$), meaning no linear directional alignment
> * **$-1$:** Opposite directions ($\theta = 180^\circ$)
>
> If either vector is a zero vector ($\|\mathbf{a}\|_2 = 0$ or $\|\mathbf{b}\|_2 = 0$), the angle/direction is undefined (division by zero), so raise a `ValueError`.

---

### 2. Intuition
* **Orientation over magnitude intuition:** Imagine two arrows starting from the origin. Cosine similarity ignores how long the arrows are and only asks: *"Are they pointing in the same general direction?"* Scaling a vector by a positive constant (like multiplying all entries by `10`) changes its $L_2$ norm, not its direction, so cosine similarity stays unchanged.
* **ML / Feature intuition:** In machine learning, cosine similarity is ubiquitous for comparing **embeddings**—such as sentence/document embeddings in semantic search and RAG, user/item vectors in recommendation systems, or high-dimensional feature vectors where vector norm can vary with length or frequency.

---

### 3. Approach / Algorithm Behind It
1. **Validate dimensions and non-empty inputs:** Check that `len(vec1) == len(vec2)` and both are non-empty ($n > 0$). Raise a `ValueError` if dimensions don't match or inputs are empty.
2. **Compute dot product and individual $L_2$ norms:**
   * You can either compute `dot = vec_dot(vec1, vec2)`, `norm1 = vec_l2_eud(vec1)`, and `norm2 = vec_l2_eud(vec2)` using your earlier helpers, **or** compute `dot_sum`, `norm1_sq`, and `norm2_sq` in a single pass over `range(n)` and take square roots at the end.
3. **Guard against zero norms:** If `norm1 == 0` or `norm2 == 0` (or `<= 1e-12`), raise a `ValueError` because cosine similarity is undefined for zero vectors.
4. **Compute and return the similarity score:** Return `dot / (norm1 * norm2)`.

* **Time Complexity:** $O(n)$, linear in the vector dimension $n$.
* **Space Complexity:** $O(1)$ auxiliary space, storing only a few scalar values.

```python
import math
def vec_dot(vec1,vec2):
  if len(vec1)!=len(vec2):
    raise ValueError("Dimensions of the two vectors doesnot match.")
  if not vec1:
    raise ValueError("Vectors cannot be empty.")

  n = len(vec1)
  dot_sum =0
  for i in range(n):
    dot_sum+=(vec1[i]*vec2[i])

  return dot_sum

def vec_l2_eud(vec):
  n = len(vec)
  l2_sum = 0
  for i in range(n):
    l2_sum += (vec[i]**2)

  return math.sqrt(l2_sum)

def vec_normalize(vec):
  if not vec:
    raise ValueError("Vectors cannot be empty")

  norm = vec_l2_eud(vec)
  if (norm == 0) or (norm <=1e-12):
    raise ValueError("Zero vector cannot be normalized.")

  n = len(vec)
  normalize = [1]*n
  for i in range(n):
    normalize[i] = vec[i]/norm

  return normalize


def vec_cosine(vec1,vec2):
  norm1 = vec_l2_eud(vec1)
  norm2 = vec_l2_eud(vec2)
  dot_sum =vec_dot(vec1,vec2)

  if (norm1 ==0) or (norm2==0) or (norm1 <= 1e-12) or (norm2 <= 1e-12):
    raise ValueError("Cosine simalarity is undefined for zero vectors")

  return (dot_sum/(norm1*norm2))

print(vec_cosine([1,2,3,4,5],[7,8,9,4,5]))
```

```python
import math


def vec_cosine_short(vec1, vec2):
  if not vec1 or len(vec1) != len(vec2):
    raise ValueError(
        "Vectors must be non-empty and have the same dimensions."
    )

  dot = norm1_sq = norm2_sq = 0
  for x, y in zip(vec1, vec2):
    dot += x * y
    norm1_sq += x * x
    norm2_sq += y * y

  denom = math.sqrt(norm1_sq) * math.sqrt(norm2_sq)
  if denom <= 1e-12:
    raise ValueError("Cosine similarity is undefined for zero vectors.")

  return dot / denom

print(vec_cosine_short([1,2,3,4,5],[7,8,9,4,5]))
```

```python
def np_vec_cosine(vec1, vec2):
  a = np.asarray(vec1, dtype=float)
  b = np.asarray(vec2, dtype=float)
  if a.ndim != 1 or b.ndim != 1 or a.size == 0 or a.shape != b.shape:
    raise ValueError("Inputs must be non-empty 1D vectors of equal length.")
  n1 = np.linalg.norm(a, ord=2)
  n2 = np.linalg.norm(b, ord=2)
  if n1 <= 1e-12 or n2 <= 1e-12:
    raise ValueError("Cosine similarity is undefined for zero vectors.")
  return float(np.dot(a, b) / (n1 * n2))

print(np_vec_cosine([1,2,3,4,5],[7,8,9,4,5]))
```
