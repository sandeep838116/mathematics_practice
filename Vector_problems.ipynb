# Vector Addition

### 1. Idea (Definition)
Vector addition combines two vectors of the same dimension (same number of elements) to produce a new vector of that same dimension. Each component of the resulting vector is the sum of the corresponding components of the input vectors.
---
### 2. Approach / Algorithm Behind It :
- a. Validate inputs: Check that both vectors are non-empty and have equal length $n$.
- b. If their dimensions differ, the operation is undefined—flag a dimension mismatch error before doing any math.
- c. Allocate output space: Prepare a result container of size $n$ to hold the summed components.
- d. Traverse index by index: Loop through index $i$ from $0$ to $n-1$ (or $1$ to $n$ in mathematical notation), add the $i$-th element of the first vector to the $i$-th element of the second vector, and store the sum at index $i$ of the output.
- e. Return the resulting vector: Once all $n$ positions are processed, return the new vector.
- Time Complexity: $O(n)$, because each of the $n$ components is visited once.
- Space Complexity: $O(n)$ for a new output vector (or $O(1)$ auxiliary space if updating one vector in place).

```python
def vec_add(vec1, vec2):
    if len(vec1) != len(vec2):
        raise ValueError("Dimensions of the two vectors are different.")
    if not vec1:
        raise ValueError("Vectors cannot be empty.")

    n = len(vec1)
    res = [0] * n
    for i in range(n):
        res[i] = vec1[i] + vec2[i]

    return res
```

```python
print(vec_add([1,2,3,4,5],[6,7,8,9,5]))
```

# Topic 2: Vector Subtraction

### 1. Idea (Definition)
Vector subtraction takes two vectors of the **same dimension** ($n$) and produces a new vector of dimension $n$ where each element is the difference between the corresponding components of the first vector and the second vector ($\mathbf{a} - \mathbf{b}$).

> **Note:** Order matters! Unlike vector addition, vector subtraction is **not commutative** ($\mathbf{a} - \mathbf{b} \neq \mathbf{b} - \mathbf{a}$ in general; instead, $\mathbf{b} - \mathbf{a} = -(\mathbf{a} - \mathbf{b})$).

---

### 2. Approach / Algorithm Behind It
1. **Validate dimensions:** Check that both vectors have equal length $n$. If `len(vec1) != len(vec2)`, stop and raise a dimension mismatch error.
2. **Validate non-empty inputs:** Ensure the vectors are non-empty (`n > 0`).
3. **Allocate output container:** Create a result list/array of size $n$.
4. **Iterate and compute difference:** Loop index $i$ from $0$ to $n - 1$, compute `vec1[i] - vec2[i]`, and store it at index $i$ of the result.
5. **Return the result:** Return the resulting difference vector.

* **Time Complexity:** $O(n)$, visiting each element once.
* **Space Complexity:** $O(n)$ for a new output vector (or $O(1)$ auxiliary space if mutating in place).

```python
def vec_sub(vec1,vec2):
  if len(vec1) != len(vec2):
    raise ValueError("Dimensions of both vector does not match.")
  if not vec1:
    raise ValueError("Vectors cann't be empty.")
  n = len(vec1)
  res = [0]*n
  for i in range(n):
    res[i] = vec1[i]-vec2[i]
  return res


print(vec_sub([9,8,7,6],[1,2,3,4]))
```

### Topic 3: Scalar Multiplication

### 1. Idea (Definition)
Scalar multiplication takes a single vector $\mathbf{a}$ of dimension $n$ and a single number (scalar) $c$, producing a new vector of dimension $n$ where every element of $\mathbf{a}$ is multiplied by $c$.


---

### 2. Intuition
* **Geometric intuition:** Multiplying a vector by a scalar stretches, shrinks, or flips its arrow without changing the line along which it points:
  * If $|c| > 1$, the vector stretches longer.
  * If $0 < |c| < 1$, the vector shrinks shorter.
  * If $c < 0$, the vector flips to the **exact opposite direction**.
  * If $c = 0$, every component becomes $0$ (the zero vector).
* **ML / Feature intuition:** Scaling a vector uniformly adjusts its overall strength or step size—such as multiplying a gradient vector by a **learning rate** ($\alpha \cdot \nabla L$) during gradient descent, or scaling a feature/embedding vector by a weight.

---

### 3. Approach / Algorithm Behind It
1. **Validate input vector:** Check that the vector is non-empty (`n > 0`). (If `n == 0`, decide whether to raise a `ValueError` or return `[]`—raising a `ValueError` keeps your convention consistent.)
2. **Allocate output space:** Create a result list/array of length $n$.
3. **Iterate and scale each component:** Loop index $i$ from $0$ to $n - 1$, compute `scalar * vec[i]`, and store it at `res[i]`.
4. **Return the scaled vector:** Return `res`.

* **Time Complexity:** $O(n)$, visiting each of the $n$ components once.
* **Space Complexity:** $O(n)$ for a new vector (or $O(1)$ auxiliary space if scaling in place).

```python
def vec_scalar(vec,num):
  if not vec:
    raise ValueError("Vector cannot be empty.")

  n = len(vec)
  res =[1]*n
  for i in range(n):
    res[i]=vec[i]*num
  return res

print(vec_scalar([1,2,3,4,5],9))
```

# Topic 4: Dot Product (Inner Product)

### 1. Idea (Definition)
The dot product (also called the **inner product**) takes two vectors of the **same dimension** ($n$) and returns a **single scalar number** (not a vector). It is computed by multiplying corresponding components of the two vectors and summing all those products together:

$$\mathbf{a} \cdot \mathbf{b} = \sum_{i=0}^{n-1} a_i b_i = a_0 b_0 + a_1 b_1 + \dots + a_{n-1} b_{n-1}$$

> **Note:** Unlike vector addition or subtraction, the output is a **scalar**, not a vector of length $n$. Also, the dot product is **commutative** ($\mathbf{a} \cdot \mathbf{b} = \mathbf{b} \cdot \mathbf{a}$).

---

### 2. Intuition
* **Alignment & Direction intuition:** The dot product measures how much two vectors point in the **same general direction**, scaled by their lengths ($\mathbf{a} \cdot \mathbf{b} = \|\mathbf{a}\| \|\mathbf{b}\| \cos\theta$):
  * **Positive ($> 0$):** Angle between them is acute ($< 90^\circ$); they generally align.
  * **Zero ($= 0$):** Vectors are **orthogonal** (perpendicular, $90^\circ$ apart) or one is a zero vector—no alignment along each other.
  * **Negative ($< 0$):** Angle between them is obtuse ($> 90^\circ$); they point in opposing directions.
* **ML / Feature intuition:** In machine learning, the dot product is everywhere—computing a neuron's pre-activation ($\mathbf{w} \cdot \mathbf{x} + b$), scoring query-key similarity in Transformer **self-attention**, and measuring raw unnormalized similarity between embedding vectors.

---

### 3. Approach / Algorithm Behind It
1. **Validate dimensions:** Check that both vectors have equal length $n$. If `len(vec1) != len(vec2)`, raise a dimension mismatch error.
2. **Validate non-empty inputs:** Ensure `n > 0` (if empty, raise a `ValueError`).
3. **Initialize an accumulator:** Set a running sum variable (e.g., `dot_sum = 0` or `0.0`) before the loop.
4. **Iterate and accumulate products:** Loop index $i$ from $0$ to $n - 1$, multiply `vec1[i] * vec2[i]`, and add that product to `dot_sum`.
5. **Return the scalar total:** Return `dot_sum` after the loop finishes.

* **Time Complexity:** $O(n)$, visiting each pair of elements once.
* **Space Complexity:** $O(1)$ auxiliary space, since you only accumulate a single scalar total rather than allocating a new array.

```python
def vec_dot(vec1,vec2):
  if len(vec1)!=len(vec2):
    raise ValueError("Dimensions of the two vectors doesnot match.")
  if not vec1:
    raise ValueError("Vectors cannot be empty.")

  n = len(vec1)
  dot_sum =0
  for i in range(n):
    dot_sum+=(vec1[i]*vec2[i])

  return dot_sum

print(vec_dot([1,2,3,4,5],[7,8,9,4,6]))
```

# Topic 5: Cross Product

### 1. Idea (Definition)
The cross product ($\mathbf{a} \times \mathbf{b}$) is a binary operation defined specifically on **two 3-dimensional vectors** ($\mathbf{a} = [a_x, a_y, a_z]$ and $\mathbf{b} = [b_x, b_y, b_z]$). Unlike the dot product (which returns a scalar), the cross product returns a **new 3D vector** that is perpendicular (orthogonal) to both input vectors:

$$\mathbf{a} \times \mathbf{b} = \begin{bmatrix} a_y b_z - a_z b_y \\ a_z b_x - a_x b_z \\ a_x b_y - a_y b_x \end{bmatrix}$$

> **Note:** The cross product is **anti-commutative** (skew-symmetric): $\mathbf{b} \times \mathbf{a} = -(\mathbf{a} \times \mathbf{b})$. Flipping the order flips the direction of the resulting vector. Also, unlike the dot product, it is **not associative** in general ($\mathbf{a} \times (\mathbf{b} \times \mathbf{c}) \neq (\mathbf{a} \times \mathbf{b}) \times \mathbf{c}$).

---

### 2. Intuition
* **Geometric intuition:** Imagine two arrows lying flat on a table—that defines a plane. The cross product points straight **up or down out of that plane** (at $90^\circ$ to both arrows), following the **right-hand rule**. Its length equals $\|\mathbf{a}\|\|\mathbf{b}\|\sin\theta$, which is the **area of the parallelogram** formed by $\mathbf{a}$ and $\mathbf{b}$. If the two vectors are parallel or anti-parallel ($\theta = 0^\circ$ or $180^\circ$), $\sin\theta = 0$, so the cross product is the zero vector $[0, 0, 0]$.
* **ML / Engineering intuition:** While high-dimensional ML mostly uses dot products and matrix multiplications rather than 3D cross products, cross products are heavily used in **3D computer graphics, robotics, physics simulations, and spatial geometry**—for instance, computing surface normals of 3D triangles, checking orientation/winding order, or finding torque ($\boldsymbol{\tau} = \mathbf{r} \times \mathbf{F}$).

---

### 3. Approach / Algorithm Behind It
1. **Validate dimensions strictly:** Check that both vectors have length **exactly `3`** (`len(vec1) == 3` and `len(vec2) == 3`). If either is empty or has any dimension other than 3, raise a `ValueError`.
2. **Unpack or index the three components:** Let the components be $(a_0, a_1, a_2)$ for `vec1` and $(b_0, b_1, b_2)$ for `vec2`.
3. **Compute the three cross-product terms:**
   * `x = a_1 * b_2 - a_2 * b_1`
   * `y = a_2 * b_0 - a_0 * b_2`
   * `z = a_0 * b_1 - a_1 * b_0`
4. **Return the 3D result vector:** Return `[x, y, z]`.

* **Time Complexity:** $O(1)$, since the dimension is fixed at $3$ (a constant number of multiplications and subtractions).
* **Space Complexity:** $O(1)$ auxiliary space, returning a fixed 3-element vector.

```python
def vec_cross(vec1,vec2):
  if (len(vec1)!=3) or (len(vec2)!=3):
    raise ValueError("Dimension of the vectors must be 3.")
  if not vec1:
    raise ValueError("Vectors cannot be empty.")

  a_1,a_2,a_3 =vec1
  b_1,b_2,b_3 =vec2

  z = (a_1*b_2)-(a_2*b_1)
  y=(a_3*b_1)-(a_1*b_3)
  x = (a_2*b_3)-(a_3*b_2)
  return [x,y,z]

print(vec_cross([1,5,9],[7,5,3]))
```

# Topic 6: Hadamard Product (Element-wise Product)

### 1. Idea (Definition)
The Hadamard product ($\mathbf{a} \odot \mathbf{b}$) takes two vectors of the **same dimension** ($n$) and produces a **new vector of the same dimension $n$** where each element is the product of the corresponding elements of the two input vectors:

$$(\mathbf{a} \odot \mathbf{b})_i = a_i b_i \quad \text{for } i = 0, 1, \dots, n-1$$

> **Note:** Unlike the **dot product** (which multiplies corresponding elements and sums them into a single scalar), the Hadamard product **keeps each product separate** inside a vector of length $n$. Like vector addition and the dot product, it is **commutative** ($\mathbf{a} \odot \mathbf{b} = \mathbf{b} \odot \mathbf{a}$).

---

### 2. Intuition
* **Component-wise modulation intuition:** Think of one vector as a signal and the second vector as a **filter, mask, or gain control** applied to each dimension independently. Where the mask/weight is `0`, that component gets zeroed out; where it is `> 1`, that component is amplified; where it is negative, that component flips sign.
* **ML / Feature intuition:** In neural networks and ML, element-wise multiplication is heavily used for **gating mechanisms** (such as forget/input/output gates in LSTMs and GRUs, or SwiGLU/GLU activations in modern LLMs), **feature weighting/attention masks**, and **gradient scaling** during backpropagation.

---

### 3. Approach / Algorithm Behind It
1. **Validate dimensions:** Check that `len(vec1) == len(vec2)`. If they differ, raise a `ValueError` for dimension mismatch.
2. **Validate non-empty inputs:** Ensure `n > 0` (raise a `ValueError` if empty).
3. **Allocate output space:** Create a result list/array of size $n$.
4. **Iterate and multiply element-wise:** Loop index $i$ from $0$ to $n - 1$, compute `vec1[i] * vec2[i]`, and store it at `res[i]`.
5. **Return the resulting vector:** Return `res`.

* **Time Complexity:** $O(n)$, visiting each of the $n$ index positions once.
* **Space Complexity:** $O(n)$ for the output vector of length $n$.

```python
def vec_hadamard(vec1,vec2):
  if len(vec1)!=len(vec2):
    raise ValueError("Dimension of the vectors are not same.")
  if not vec1:
    raise ValueError("Vectors cannot be empty.")
  n=len(vec1)
  res=[1]*n
  for i in range(n):
    res[i]= vec1[i]*vec2[i]

  return res

print(vec_hadamard([7,4,1],[8,5,2]))
```

# Topic 7: Outer Product

### 1. Idea (Definition)
The outer product ($\mathbf{a} \otimes \mathbf{b}$ or $\mathbf{a}\mathbf{b}^T$) takes **two 1D vectors**—say $\mathbf{a}$ of length $m$ and $\mathbf{b}$ of length $n$—and produces a **2D matrix of shape $(m, n)$** ($m$ rows and $n$ columns). Each entry at row $i$ and column $j$ is the product of the $i$-th element of $\mathbf{a}$ and the $j$-th element of $\mathbf{b}$:

$$(\mathbf{a} \otimes \mathbf{b})_{i, j} = a_i \cdot b_j \quad \text{for } 0 \le i < m,\; 0 \le j < n$$

> **Note:** Unlike vector addition, subtraction, dot product, or Hadamard product, the outer product **does NOT require $\mathbf{a}$ and $\mathbf{b}$ to have the same length** ($m$ and $n$ can be different!), and its result is a **2D grid (matrix)** rather than a 1D vector or scalar. Also, $\mathbf{a} \otimes \mathbf{b}$ is the transpose of $\mathbf{b} \otimes \mathbf{a}$ (not equal in general).

---

### 2. Intuition
* **Pairwise interaction grid intuition:** Think of row $i$ as scaling the entire vector $\mathbf{b}$ by $a_i$ (or equivalently, column $j$ as scaling vector $\mathbf{a}$ by $b_j$). Every cell $(i, j)$ captures the pairwise interaction between feature/coordinate $i$ of the first vector and feature/coordinate $j$ of the second vector.
* **ML / Feature intuition:** In machine learning, outer products build **rank-1 matrices**, covariance/gram-style outer terms, and **bilinear / cross-feature interactions** (such as combining user features and item features into an interaction matrix, or forming weight updates from activation and error vectors).

---

### 3. Approach / Algorithm Behind It
1. **Validate non-empty inputs:** Check that both `vec1` (length $m$) and `vec2` (length $n$) are non-empty (`m > 0` and `n > 0`). If either is empty, raise a `ValueError`.
2. **Determine matrix dimensions:** Let `m = len(vec1)` (number of rows) and `n = len(vec2)` (number of columns).
3. **Allocate a 2D result grid:** Create an $m \times n$ matrix (a list of $m$ rows, where each row has $n$ elements—be careful to create a fresh row per `i` rather than sharing the same row reference across all rows!).
4. **Nested iteration over rows and columns:**
   * Outer loop: `i` from `0` to `m - 1` (selecting `vec1[i]`).
   * Inner loop: `j` from `0` to `n - 1` (selecting `vec2[j]`), computing `vec1[i] * vec2[j]`, and storing it at position `[i][j]`.
5. **Return the 2D matrix:** Return the resulting $m \times n$ list of lists.

* **Time Complexity:** $O(m \cdot n)$, because every pair $(i, j)$ out of $m \times n$ entries is computed once.
* **Space Complexity:** $O(m \cdot n)$ to store the resulting $m \times n$ matrix.

```python
def vec_outer(vec1,vec2):
  if not vec1 or not vec2:
    raise ValueError("Vectors cannot be empty.")

  m = len(vec1)
  n = len(vec2)
  res = [[0 for _ in range(n)] for _ in range(m)]

  for i in range(m):
    for j in range(n):
      res[i][j]=vec1[i]*vec2[j]

  return res

print(vec_outer([1,2,3,4,5],[8,5,2]))
```
